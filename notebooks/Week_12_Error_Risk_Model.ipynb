```python
from pathlib import Path
from datetime import datetime, timezone
import json
import pickle
import sys
import zipfile
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    brier_score_loss,
    log_loss,
    confusion_matrix,
    roc_curve,
)

try:
    from IPython.display import display
except ImportError:
    def display(obj):
        print(obj)

warnings.filterwarnings("ignore", category=FutureWarning)

# ------------------------------------------------------------
# 1. Week 12 configuration
# ------------------------------------------------------------

PROJECT_TITLE = "Against All Odds"
WEEK_NUMBER = 12
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

CORE_FEATURES = [
    "ELO_DIFF",
    "WIN_PCT_DIFF",
    "FORM_DIFF",
    "POINT_GOAL_DIFF",
    "OFFENSE_DIFF",
    "DEFENSE_DIFF",
    "REST_DIFF",
    "SCHEDULE_STRENGTH_DIFF",
]

FRAGILITY_FEATURES = [
    "ELO_DIFF",
    "FORM_DIFF",
    "OFFENSE_DIFF",
    "DEFENSE_DIFF",
    "REST_DIFF",
]

MODEL_READY_INPUTS = {
    "NBA": "nba_model_ready_2021_22_to_2025_26.csv",
    "NFL": "nfl_model_ready_2021_to_2025.csv",
    "Premier League": "premier_league_model_ready_2021_22_to_2025_26.csv",
}

WEEK9_SELECTION_FILE = "week9_best_primary_model_by_sport.csv"

GB_PARAMS = {
    "n_estimators": 150,
    "learning_rate": 0.05,
    "max_depth": 3,
    "min_samples_leaf": 10,
    "subsample": 0.90,
    "random_state": RANDOM_SEED,
}

# Same simple standardized perturbation idea as Week 11.
STEP_SD = 0.25
MAX_FLIP_SD = 4.0

PERTURBATION_STEPS = np.arange(
    STEP_SD,
    MAX_FLIP_SD + STEP_SD / 2,
    STEP_SD,
)

# Potential error-risk predictors from the capstone draft.
# HOME_ADVANTAGE may be constant in the current model-ready data,
# so zero-variance predictors are dropped automatically.
RISK_FEATURE_CANDIDATES = [
    "ORIGINAL_WIN_PROBABILITY",
    "FRAGILITY_SCORE",
    "ABS_ELO_DIFF",
    "ABS_TEAM_STRENGTH_DIFF",
    "ABS_FORM_DIFF",
    "MODEL_DISAGREEMENT_COUNT",
    "ABS_REST_DIFF",
    "HOME_ADVANTAGE",
    "PRIMARY_PREDICTS_HOME",
]

print("=" * 78)
print(f"{PROJECT_TITLE} — WEEK {WEEK_NUMBER}")
print("PREDICTION ERROR-RISK MODEL")
print("=" * 78)
print("Primary model fit split: TRAIN")
print("Error-risk model fit split: VALIDATION")
print("Final error-risk evaluation split: TEST")


# ------------------------------------------------------------
# 2. Project folders
# ------------------------------------------------------------

def running_in_colab():
    return "google.colab" in sys.modules


def choose_project_root():
    cwd = Path.cwd().resolve()

    candidates = [
        cwd if cwd.name.lower().replace("_", "-") in {
            "againstallodds",
            "against-all-odds",
        } else None,
        Path("/content/againstallodds"),
        Path("/content/Against-All-Odds"),
        Path("/content/drive/MyDrive/againstallodds"),
        Path("/content/drive/MyDrive/Against-All-Odds"),
    ]

    for candidate in candidates:
        if candidate is not None and candidate.exists():
            return candidate

    if running_in_colab():
        return Path("/content/againstallodds")

    return cwd / "againstallodds"


PROJECT_ROOT = choose_project_root()

MODEL_READY_DIR = PROJECT_ROOT / "data" / "model_ready"
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = PROJECT_ROOT / "figures" / "week12"
DOCS_DIR = PROJECT_ROOT / "docs"
MODELS_DIR = PROJECT_ROOT / "models" / "week12"

for folder in [
    MODEL_READY_DIR,
    RESULTS_DIR,
    FIGURES_DIR,
    DOCS_DIR,
    MODELS_DIR,
]:
    folder.mkdir(
        parents=True,
        exist_ok=True,
    )

print("\nProject root:", PROJECT_ROOT)
print("RESULTS_DIR:", RESULTS_DIR)
print("FIGURES_DIR:", FIGURES_DIR)
print("DOCS_DIR:", DOCS_DIR)
print("MODELS_DIR:", MODELS_DIR)


# ------------------------------------------------------------
# 3. Locate required inputs
# ------------------------------------------------------------

def search_locations():
    locations = [
        PROJECT_ROOT,
        MODEL_READY_DIR,
        RESULTS_DIR,
        Path("/content"),
        Path("/content/week12_inputs"),
        Path.cwd().resolve(),
    ]

    for extra in [
        Path("/content/drive/MyDrive/againstallodds"),
        Path("/content/drive/MyDrive/Against-All-Odds"),
    ]:
        if extra.exists():
            locations.append(extra)

    return list(
        dict.fromkeys(
            path
            for path in locations
            if path.exists()
        )
    )


def find_exact_file(filename):
    for root in search_locations():
        direct = root / filename

        if direct.exists():
            return direct

    for root in search_locations():
        try:
            matches = list(
                root.rglob(filename)
            )
        except (PermissionError, OSError):
            matches = []

        if matches:
            return matches[0]

    return None


def locate_required_files():
    required_names = (
        list(MODEL_READY_INPUTS.values())
        + [WEEK9_SELECTION_FILE]
    )

    found = {
        name: find_exact_file(name)
        for name in required_names
    }

    missing = [
        name
        for name, path in found.items()
        if path is None
    ]

    if missing and running_in_colab():
        print("\nMissing files needed for Week 12:")

        for name in missing:
            print(" -", name)

        print(
            "\nUpload the missing CSV files "
            "or a ZIP containing them."
        )

        from google.colab import files

        uploaded = files.upload()

        upload_dir = Path("/content/week12_inputs")
        upload_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        for name, blob in uploaded.items():
            destination = (
                upload_dir
                / Path(name).name
            )

            destination.write_bytes(blob)

            if zipfile.is_zipfile(destination):
                with zipfile.ZipFile(
                    destination,
                    "r",
                ) as archive:
                    archive.extractall(
                        upload_dir
                    )

        found = {
            name: find_exact_file(name)
            for name in required_names
        }

    still_missing = [
        name
        for name, path in found.items()
        if path is None
    ]

    if still_missing:
        raise FileNotFoundError(
            "Week 12 cannot start because these files are missing:\n"
            + "\n".join(
                f" - {name}"
                for name in still_missing
            )
            + "\n\nRun Week 6 and Week 9 first, "
              "or upload the missing files."
        )

    return found


LOCATED_FILES = locate_required_files()

print("\n✅ Required Week 12 files located:")

for name, path in LOCATED_FILES.items():
    print(" -", name, "->", path)


# ------------------------------------------------------------
# 4. Load and validate model-ready datasets
# ------------------------------------------------------------

datasets = {}

required_columns = (
    [
        "game_id",
        "game_date",
        "data_split",
        "HOME_WIN",
    ]
    + CORE_FEATURES
)

for sport, filename in MODEL_READY_INPUTS.items():
    path = LOCATED_FILES[filename]

    df = pd.read_csv(
        path,
        dtype={"game_id": "string"},
        parse_dates=["game_date"],
    )

    df["data_split"] = (
        df["data_split"]
        .astype("string")
        .str.upper()
        .str.strip()
    )

    df["HOME_WIN"] = pd.to_numeric(
        df["HOME_WIN"],
        errors="coerce",
    )

    missing = [
        col
        for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"{sport} is missing required Week 12 columns: {missing}"
        )

    for feature in CORE_FEATURES:
        df[feature] = pd.to_numeric(
            df[feature],
            errors="coerce",
        )

    # HOME_ADVANTAGE was created in Week 6. If an older
    # model-ready file does not contain it, use a home-context
    # indicator of 1 for every home-perspective game row.
    if "HOME_ADVANTAGE" not in df.columns:
        df["HOME_ADVANTAGE"] = 1.0

    df["HOME_ADVANTAGE"] = pd.to_numeric(
        df["HOME_ADVANTAGE"],
        errors="coerce",
    ).fillna(1.0)

    df = (
        df
        .dropna(
            subset=CORE_FEATURES + ["HOME_WIN"]
        )
        .sort_values(
            ["game_date", "game_id"],
            kind="stable",
        )
        .reset_index(drop=True)
    )

    if not df["HOME_WIN"].isin([0, 1]).all():
        raise ValueError(
            f"{sport}: HOME_WIN must contain only 0 or 1."
        )

    datasets[sport] = df

    print(
        f"{sport}: {len(df):,} rows | "
        f"TRAIN={int(df['data_split'].eq('TRAIN').sum()):,} | "
        f"VALIDATION={int(df['data_split'].eq('VALIDATION').sum()):,} | "
        f"TEST={int(df['data_split'].eq('TEST').sum()):,}"
    )


# ------------------------------------------------------------
# 5. Load Week 9 selected primary models
# ------------------------------------------------------------

selection_path = LOCATED_FILES[
    WEEK9_SELECTION_FILE
]

week9_selection = pd.read_csv(
    selection_path
)

required_selection_columns = [
    "sport",
    "model",
]

missing_selection = [
    col
    for col in required_selection_columns
    if col not in week9_selection.columns
]

if missing_selection:
    raise ValueError(
        "week9_best_primary_model_by_sport.csv "
        f"is missing required columns: {missing_selection}"
    )

week9_selection["sport"] = (
    week9_selection["sport"]
    .astype("string")
    .str.strip()
)

week9_selection["model"] = (
    week9_selection["model"]
    .astype("string")
    .str.strip()
)

PRIMARY_MODEL_BY_SPORT = {
    row.sport: row.model
    for row in week9_selection[
        ["sport", "model"]
    ].itertuples(index=False)
}

expected_sports = {
    "NBA",
    "NFL",
    "Premier League",
}

if set(PRIMARY_MODEL_BY_SPORT) != expected_sports:
    raise ValueError(
        "Week 9 model-selection file must contain "
        "NBA, NFL, and Premier League."
    )

print("\nWeek 9 selected primary models:")

for sport, model_name in PRIMARY_MODEL_BY_SPORT.items():
    print(f" - {sport}: {model_name}")


# ------------------------------------------------------------
# 6. Winner-model helpers
# ------------------------------------------------------------

def elo_probability(elo_difference):
    elo_difference = np.asarray(
        elo_difference,
        dtype=float,
    )

    return (
        1.0
        / (
            1.0
            + 10.0 ** (
                -elo_difference
                / 400.0
            )
        )
    )


def build_logistic_winner_model():
    return Pipeline([
        (
            "scaler",
            StandardScaler(),
        ),
        (
            "logistic",
            LogisticRegression(
                max_iter=5000,
                random_state=RANDOM_SEED,
            ),
        ),
    ])


def build_gradient_boosting_winner_model():
    return GradientBoostingClassifier(
        **GB_PARAMS
    )


def fit_candidate_winner_models(train):
    logistic_model = (
        build_logistic_winner_model()
    )

    logistic_model.fit(
        train[CORE_FEATURES],
        train["HOME_WIN"].astype(int),
    )

    gradient_model = (
        build_gradient_boosting_winner_model()
    )

    gradient_model.fit(
        train[CORE_FEATURES],
        train["HOME_WIN"].astype(int),
    )

    return {
        "Logistic Regression": logistic_model,
        "Gradient Boosting": gradient_model,
        "Elo": None,
    }


def candidate_probability(
    frame,
    model_name,
    fitted_models,
):
    if model_name == "Elo":
        return elo_probability(
            frame["ELO_DIFF"]
        )

    if model_name in fitted_models:
        return (
            fitted_models[
                model_name
            ]
            .predict_proba(
                frame[CORE_FEATURES]
            )[:, 1]
        )

    raise ValueError(
        f"Unsupported winner model: {model_name}"
    )


# ------------------------------------------------------------
# 7. Week 11-style fragility for any development split
# ------------------------------------------------------------

def fragility_category(score):
    if score >= 75:
        return "Extremely Fragile"

    if score >= 50:
        return "Fragile"

    if score >= 25:
        return "Moderate"

    return "Stable"


def compute_fragility(
    frame,
    train,
    primary_model_name,
    fitted_models,
):
    """
    Compute the Week 11-style single-feature winner-flip distance.

    For Week 12, the original raw primary-model probability is used
    so VALIDATION can safely train the error-risk model without
    borrowing its outcomes through probability calibration.
    """

    original_probability = candidate_probability(
        frame=frame,
        model_name=primary_model_name,
        fitted_models=fitted_models,
    )

    original_prediction = (
        original_probability >= 0.50
    ).astype(int)

    n_games = len(frame)

    feature_scale = {}
    feature_min = {}
    feature_max = {}

    for feature in FRAGILITY_FEATURES:
        scale = float(
            train[feature].std(
                ddof=0
            )
        )

        if (
            not np.isfinite(scale)
            or scale <= 0
        ):
            scale = 1.0

        feature_scale[feature] = scale
        feature_min[feature] = float(
            train[feature].min()
        )
        feature_max[feature] = float(
            train[feature].max()
        )

    per_feature_frames = []

    for feature in FRAGILITY_FEATURES:
        scale = feature_scale[feature]

        best_distance = np.full(
            n_games,
            np.inf,
            dtype=float,
        )

        best_direction = np.full(
            n_games,
            "",
            dtype=object,
        )

        best_probability = np.full(
            n_games,
            np.nan,
            dtype=float,
        )

        original_value = (
            frame[feature]
            .to_numpy(
                dtype=float
            )
        )

        for step_sd in PERTURBATION_STEPS:
            unresolved = (
                ~np.isfinite(
                    best_distance
                )
            )

            if not unresolved.any():
                break

            for sign, direction_name in [
                (-1.0, "Decrease"),
                (1.0, "Increase"),
            ]:
                unresolved = (
                    ~np.isfinite(
                        best_distance
                    )
                )

                if not unresolved.any():
                    break

                perturbed = frame.copy()

                proposed_value = (
                    original_value
                    + (
                        sign
                        * step_sd
                        * scale
                    )
                )

                proposed_value = np.clip(
                    proposed_value,
                    feature_min[feature],
                    feature_max[feature],
                )

                perturbed[feature] = (
                    proposed_value
                )

                perturbed_probability = (
                    candidate_probability(
                        frame=perturbed,
                        model_name=primary_model_name,
                        fitted_models=fitted_models,
                    )
                )

                perturbed_prediction = (
                    perturbed_probability
                    >= 0.50
                ).astype(int)

                actual_distance = (
                    np.abs(
                        proposed_value
                        - original_value
                    )
                    / scale
                )

                new_flip = (
                    unresolved
                    & (
                        perturbed_prediction
                        != original_prediction
                    )
                    & (
                        actual_distance > 0
                    )
                )

                if new_flip.any():
                    best_distance[
                        new_flip
                    ] = actual_distance[
                        new_flip
                    ]

                    best_direction[
                        new_flip
                    ] = direction_name

                    best_probability[
                        new_flip
                    ] = perturbed_probability[
                        new_flip
                    ]

        per_feature_frames.append(
            pd.DataFrame({
                "game_id": (
                    frame["game_id"]
                    .astype("string")
                    .to_numpy()
                ),
                "feature": feature,
                "flip_found": (
                    np.isfinite(
                        best_distance
                    )
                ),
                "flip_distance_sd": np.where(
                    np.isfinite(
                        best_distance
                    ),
                    best_distance,
                    np.nan,
                ),
                "flip_direction": best_direction,
                "probability_at_flip": (
                    best_probability
                ),
            })
        )

    feature_flip_results = pd.concat(
        per_feature_frames,
        ignore_index=True,
    )

    finite_only = (
        feature_flip_results[
            feature_flip_results[
                "flip_found"
            ]
        ]
        .sort_values(
            [
                "game_id",
                "flip_distance_sd",
                "feature",
            ]
        )
    )

    if finite_only.empty:
        first_flip = pd.DataFrame({
            "game_id": pd.Series(
                dtype="string"
            ),
            "FRAGILITY_DRIVER": pd.Series(
                dtype="string"
            ),
            "MIN_FLIP_DISTANCE_SD": pd.Series(
                dtype=float
            ),
        })

    else:
        first_flip = (
            finite_only
            .groupby(
                "game_id",
                as_index=False,
            )
            .first()
            .rename(
                columns={
                    "feature": "FRAGILITY_DRIVER",
                    "flip_distance_sd": (
                        "MIN_FLIP_DISTANCE_SD"
                    ),
                }
            )
        )

    result = pd.DataFrame({
        "game_id": (
            frame["game_id"]
            .astype("string")
            .to_numpy()
        ),
        "ORIGINAL_HOME_WIN_PROBABILITY": (
            original_probability
        ),
        "ORIGINAL_PREDICTION": (
            original_prediction
        ),
    })

    result = result.merge(
        first_flip[
            [
                "game_id",
                "FRAGILITY_DRIVER",
                "MIN_FLIP_DISTANCE_SD",
            ]
        ],
        on="game_id",
        how="left",
    )

    result[
        "FLIP_FOUND_WITHIN_RANGE"
    ] = (
        result[
            "MIN_FLIP_DISTANCE_SD"
        ].notna()
    )

    capped_distance = (
        result[
            "MIN_FLIP_DISTANCE_SD"
        ]
        .fillna(
            MAX_FLIP_SD
        )
        .clip(
            lower=0,
            upper=MAX_FLIP_SD,
        )
    )

    result[
        "FRAGILITY_SCORE"
    ] = (
        100.0
        * (
            1.0
            - (
                capped_distance
                / MAX_FLIP_SD
            )
        )
    ).clip(
        0,
        100,
    )

    result[
        "FRAGILITY_SCORE"
    ] = (
        result[
            "FRAGILITY_SCORE"
        ].round(2)
    )

    result[
        "FRAGILITY_CATEGORY"
    ] = (
        result[
            "FRAGILITY_SCORE"
        ]
        .map(
            fragility_category
        )
    )

    return (
        result,
        feature_flip_results,
    )


# ------------------------------------------------------------
# 8. Build the second dataset for error-risk modeling
# ------------------------------------------------------------

def build_error_risk_dataset(
    sport,
    frame,
    train,
    primary_model_name,
    fitted_models,
    risk_split_name,
):
    """
    Build one historical-prediction dataset.

    PRIMARY_MODEL_CORRECT:
        1 = original winner prediction was correct
        0 = original winner prediction was wrong

    PRIMARY_MODEL_ERROR:
        1 = original winner prediction was wrong
        0 = original winner prediction was correct

    The error-risk model predicts PRIMARY_MODEL_ERROR.
    """

    fragility, feature_flip = (
        compute_fragility(
            frame=frame,
            train=train,
            primary_model_name=(
                primary_model_name
            ),
            fitted_models=fitted_models,
        )
    )

    working = frame.copy()

    working["game_id"] = (
        working["game_id"]
        .astype("string")
    )

    working = working.merge(
        fragility,
        on="game_id",
        how="left",
    )

    # Candidate-model probabilities.
    elo_prob = candidate_probability(
        frame=working,
        model_name="Elo",
        fitted_models=fitted_models,
    )

    logistic_prob = candidate_probability(
        frame=working,
        model_name="Logistic Regression",
        fitted_models=fitted_models,
    )

    gradient_prob = candidate_probability(
        frame=working,
        model_name="Gradient Boosting",
        fitted_models=fitted_models,
    )

    working[
        "ELO_HOME_WIN_PROB"
    ] = elo_prob

    working[
        "LOGISTIC_HOME_WIN_PROB"
    ] = logistic_prob

    working[
        "GRADIENT_BOOSTING_HOME_WIN_PROB"
    ] = gradient_prob

    # Original prediction confidence means the probability
    # assigned to whichever team the primary model selected.
    working[
        "ORIGINAL_WIN_PROBABILITY"
    ] = np.maximum(
        working[
            "ORIGINAL_HOME_WIN_PROBABILITY"
        ],
        1.0
        - working[
            "ORIGINAL_HOME_WIN_PROBABILITY"
        ],
    )

    candidate_predictions = np.column_stack([
        (
            elo_prob >= 0.50
        ).astype(int),
        (
            logistic_prob >= 0.50
        ).astype(int),
        (
            gradient_prob >= 0.50
        ).astype(int),
    ])

    primary_prediction = (
        working[
            "ORIGINAL_PREDICTION"
        ]
        .to_numpy(
            dtype=int
        )
    )

    # 0 means all models agree with the selected primary model.
    # 2 means both alternative models disagree with it.
    working[
        "MODEL_DISAGREEMENT_COUNT"
    ] = (
        candidate_predictions
        != primary_prediction.reshape(
            -1,
            1,
        )
    ).sum(
        axis=1
    )

    working[
        "MODEL_PROBABILITY_SPREAD"
    ] = (
        np.max(
            np.column_stack([
                elo_prob,
                logistic_prob,
                gradient_prob,
            ]),
            axis=1,
        )
        - np.min(
            np.column_stack([
                elo_prob,
                logistic_prob,
                gradient_prob,
            ]),
            axis=1,
        )
    )

    # Draft inputs converted to easy-to-read risk features.
    working[
        "ABS_ELO_DIFF"
    ] = (
        working[
            "ELO_DIFF"
        ].abs()
    )

    working[
        "ABS_TEAM_STRENGTH_DIFF"
    ] = (
        working[
            "WIN_PCT_DIFF"
        ].abs()
    )

    working[
        "ABS_FORM_DIFF"
    ] = (
        working[
            "FORM_DIFF"
        ].abs()
    )

    working[
        "ABS_REST_DIFF"
    ] = (
        working[
            "REST_DIFF"
        ].abs()
    )

    working[
        "PRIMARY_PREDICTS_HOME"
    ] = (
        working[
            "ORIGINAL_PREDICTION"
        ].astype(int)
    )

    working[
        "PRIMARY_MODEL_CORRECT"
    ] = (
        working[
            "ORIGINAL_PREDICTION"
        ].astype(int)
        == working[
            "HOME_WIN"
        ].astype(int)
    ).astype("int8")

    working[
        "PRIMARY_MODEL_ERROR"
    ] = (
        1
        - working[
            "PRIMARY_MODEL_CORRECT"
        ]
    ).astype("int8")

    working[
        "sport"
    ] = sport

    working[
        "PRIMARY_MODEL"
    ] = primary_model_name

    working[
        "ERROR_RISK_DATA_SPLIT"
    ] = risk_split_name

    # Keep useful identifiers plus all Week 12 risk variables.
    identifier_candidates = [
        "sport",
        "game_id",
        "game_date",
        "season",
        "home_team",
        "away_team",
        "home_team_abbr",
        "away_team_abbr",
        "data_split",
        "ERROR_RISK_DATA_SPLIT",
        "HOME_WIN",
        "PRIMARY_MODEL",
        "ORIGINAL_HOME_WIN_PROBABILITY",
        "ORIGINAL_WIN_PROBABILITY",
        "ORIGINAL_PREDICTION",
        "PRIMARY_MODEL_CORRECT",
        "PRIMARY_MODEL_ERROR",
        "FRAGILITY_SCORE",
        "FRAGILITY_CATEGORY",
        "FRAGILITY_DRIVER",
        "MIN_FLIP_DISTANCE_SD",
        "FLIP_FOUND_WITHIN_RANGE",
        "ELO_DIFF",
        "WIN_PCT_DIFF",
        "FORM_DIFF",
        "REST_DIFF",
        "HOME_ADVANTAGE",
        "ABS_ELO_DIFF",
        "ABS_TEAM_STRENGTH_DIFF",
        "ABS_FORM_DIFF",
        "ABS_REST_DIFF",
        "MODEL_DISAGREEMENT_COUNT",
        "MODEL_PROBABILITY_SPREAD",
        "PRIMARY_PREDICTS_HOME",
        "ELO_HOME_WIN_PROB",
        "LOGISTIC_HOME_WIN_PROB",
        "GRADIENT_BOOSTING_HOME_WIN_PROB",
    ]

    output_columns = [
        col
        for col in identifier_candidates
        if col in working.columns
    ]

    return (
        working[
            output_columns
        ].copy(),
        feature_flip,
    )


# ------------------------------------------------------------
# 9. Error-risk model helpers
# ------------------------------------------------------------

def select_risk_features(
    training_frame,
):
    """
    Remove any candidate feature that is missing or has zero
    variance in VALIDATION. This is especially useful for
    HOME_ADVANTAGE, which may be constant in the current data.
    """

    selected = []

    for feature in RISK_FEATURE_CANDIDATES:
        if feature not in training_frame.columns:
            continue

        series = pd.to_numeric(
            training_frame[feature],
            errors="coerce",
        )

        if series.notna().sum() == 0:
            continue

        if series.nunique(
            dropna=True
        ) <= 1:
            print(
                f"Skipping zero-variance risk feature: {feature}"
            )
            continue

        selected.append(feature)

    if len(selected) < 2:
        raise ValueError(
            "Too few usable Week 12 risk features."
        )

    return selected


def build_error_risk_model():
    """
    Balanced logistic regression keeps Week 12 simple and
    interpretable while accounting for prediction errors that
    may be less common than correct predictions.
    """

    return Pipeline([
        (
            "scaler",
            StandardScaler(),
        ),
        (
            "logistic",
            LogisticRegression(
                max_iter=5000,
                class_weight="balanced",
                random_state=RANDOM_SEED,
            ),
        ),
    ])


def evaluate_error_risk(
    y_true,
    probability,
):
    probability = np.clip(
        np.asarray(
            probability,
            dtype=float,
        ),
        1e-6,
        1 - 1e-6,
    )

    prediction = (
        probability >= 0.50
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1],
    ).ravel()

    return {
        "accuracy": accuracy_score(
            y_true,
            prediction,
        ),
        "precision_error": precision_score(
            y_true,
            prediction,
            zero_division=0,
        ),
        "recall_error": recall_score(
            y_true,
            prediction,
            zero_division=0,
        ),
        "f1_error": f1_score(
            y_true,
            prediction,
            zero_division=0,
        ),
        "roc_auc": roc_auc_score(
            y_true,
            probability,
        ),
        "brier_score": brier_score_loss(
            y_true,
            probability,
        ),
        "log_loss": log_loss(
            y_true,
            probability,
            labels=[0, 1],
        ),
        "true_negative": int(tn),
        "false_positive": int(fp),
        "false_negative": int(fn),
        "true_positive": int(tp),
    }


def save_confusion_matrix_figure(
    y_true,
    probability,
    sport,
):
    prediction = (
        np.asarray(
            probability
        )
        >= 0.50
    ).astype(int)

    matrix = confusion_matrix(
        y_true,
        prediction,
        labels=[0, 1],
    )

    fig, ax = plt.subplots(
        figsize=(5, 4)
    )

    image = ax.imshow(
        matrix
    )

    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])

    ax.set_xticklabels([
        "Predict Correct",
        "Predict Error",
    ])

    ax.set_yticklabels([
        "Actual Correct",
        "Actual Error",
    ])

    ax.set_xlabel(
        "Predicted Error-Risk Class"
    )

    ax.set_ylabel(
        "Actual Primary-Model Result"
    )

    ax.set_title(
        f"{sport}: Error-Risk Confusion Matrix"
    )

    for i in range(2):
        for j in range(2):
            ax.text(
                j,
                i,
                str(
                    matrix[i, j]
                ),
                ha="center",
                va="center",
            )

    fig.colorbar(
        image,
        ax=ax,
    )

    fig.tight_layout()

    safe_sport = (
        sport.lower()
        .replace(" ", "_")
    )

    path = (
        FIGURES_DIR
        / (
            f"week12_{safe_sport}_"
            "error_risk_confusion.png"
        )
    )

    fig.savefig(
        path,
        dpi=160,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

    return path


def save_roc_figure(
    y_true,
    probability,
    sport,
):
    fpr, tpr, _ = roc_curve(
        y_true,
        probability,
    )

    auc_value = roc_auc_score(
        y_true,
        probability,
    )

    fig, ax = plt.subplots(
        figsize=(6, 5)
    )

    ax.plot(
        fpr,
        tpr,
        label=(
            f"ROC-AUC = {auc_value:.3f}"
        ),
    )

    ax.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
    )

    ax.set_xlabel(
        "False Positive Rate"
    )

    ax.set_ylabel(
        "True Positive Rate"
    )

    ax.set_title(
        f"{sport}: Error-Risk ROC — TEST"
    )

    ax.legend()
    fig.tight_layout()

    safe_sport = (
        sport.lower()
        .replace(" ", "_")
    )

    path = (
        FIGURES_DIR
        / (
            f"week12_{safe_sport}_"
            "error_risk_roc.png"
        )
    )

    fig.savefig(
        path,
        dpi=160,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

    return path


# ------------------------------------------------------------
# 10. Build datasets, train, and evaluate by sport
# ------------------------------------------------------------

sport_outputs = {}
metrics_rows = []
coefficient_frames = []
all_training_frames = []
all_test_frames = []
all_fragility_feature_frames = []
figure_paths = []
model_paths = []

for sport, df in datasets.items():
    print("\n" + "=" * 72)
    print("WEEK 12 ERROR-RISK MODEL:", sport)
    print("=" * 72)

    train = df[
        df["data_split"].eq("TRAIN")
    ].copy()

    validation = df[
        df["data_split"].eq("VALIDATION")
    ].copy()

    test = df[
        df["data_split"].eq("TEST")
    ].copy()

    if train.empty:
        raise ValueError(
            f"{sport}: TRAIN split is empty."
        )

    if validation.empty:
        raise ValueError(
            f"{sport}: VALIDATION split is empty."
        )

    if test.empty:
        raise ValueError(
            f"{sport}: TEST split is empty."
        )

    if validation["HOME_WIN"].nunique() < 2:
        raise ValueError(
            f"{sport}: VALIDATION needs both HOME_WIN classes."
        )

    primary_model_name = (
        PRIMARY_MODEL_BY_SPORT[
            sport
        ]
    )

    # Fit all three Week 9 candidate winner models on TRAIN.
    fitted_winner_models = (
        fit_candidate_winner_models(
            train
        )
    )

    # Build the second historical-prediction dataset from VALIDATION.
    risk_train_df, validation_flip_details = (
        build_error_risk_dataset(
            sport=sport,
            frame=validation,
            train=train,
            primary_model_name=(
                primary_model_name
            ),
            fitted_models=(
                fitted_winner_models
            ),
            risk_split_name=(
                "ERROR_RISK_TRAIN"
            ),
        )
    )

    # Build the completely held-out risk TEST dataset.
    risk_test_df, test_flip_details = (
        build_error_risk_dataset(
            sport=sport,
            frame=test,
            train=train,
            primary_model_name=(
                primary_model_name
            ),
            fitted_models=(
                fitted_winner_models
            ),
            risk_split_name=(
                "ERROR_RISK_TEST"
            ),
        )
    )

    selected_risk_features = (
        select_risk_features(
            risk_train_df
        )
    )

    # Convert selected inputs to numeric and remove any
    # unexpected missing rows consistently.
    for feature in selected_risk_features:
        risk_train_df[feature] = pd.to_numeric(
            risk_train_df[feature],
            errors="coerce",
        )

        risk_test_df[feature] = pd.to_numeric(
            risk_test_df[feature],
            errors="coerce",
        )

    risk_train_df = (
        risk_train_df
        .dropna(
            subset=(
                selected_risk_features
                + [
                    "PRIMARY_MODEL_ERROR"
                ]
            )
        )
        .reset_index(
            drop=True
        )
    )

    risk_test_df = (
        risk_test_df
        .dropna(
            subset=(
                selected_risk_features
                + [
                    "PRIMARY_MODEL_ERROR"
                ]
            )
        )
        .reset_index(
            drop=True
        )
    )

    if risk_train_df[
        "PRIMARY_MODEL_ERROR"
    ].nunique() < 2:
        raise ValueError(
            f"{sport}: error-risk training data need "
            "both correct and incorrect primary predictions."
        )

    if risk_test_df[
        "PRIMARY_MODEL_ERROR"
    ].nunique() < 2:
        raise ValueError(
            f"{sport}: TEST data need both correct and "
            "incorrect primary predictions for ROC-AUC."
        )

    X_risk_train = (
        risk_train_df[
            selected_risk_features
        ].copy()
    )

    y_risk_train = (
        risk_train_df[
            "PRIMARY_MODEL_ERROR"
        ].astype(int)
    )

    X_risk_test = (
        risk_test_df[
            selected_risk_features
        ].copy()
    )

    y_risk_test = (
        risk_test_df[
            "PRIMARY_MODEL_ERROR"
        ].astype(int)
    )

    error_risk_model = (
        build_error_risk_model()
    )

    error_risk_model.fit(
        X_risk_train,
        y_risk_train,
    )

    test_error_probability = (
        error_risk_model
        .predict_proba(
            X_risk_test
        )[:, 1]
    )

    test_error_prediction = (
        test_error_probability
        >= 0.50
    ).astype("int8")

    risk_test_df[
        "ERROR_RISK_PROBABILITY"
    ] = test_error_probability

    risk_test_df[
        "ERROR_RISK_PREDICTION"
    ] = test_error_prediction

    risk_test_df[
        "ERROR_RISK_THRESHOLD"
    ] = 0.50

    metrics = evaluate_error_risk(
        y_true=y_risk_test,
        probability=(
            test_error_probability
        ),
    )

    metrics_rows.append({
        "sport": sport,
        "model": (
            "Balanced Logistic Regression"
        ),
        "primary_winner_model": (
            primary_model_name
        ),
        "risk_training_split": (
            "VALIDATION"
        ),
        "risk_evaluation_split": (
            "TEST"
        ),
        "training_predictions": int(
            len(
                risk_train_df
            )
        ),
        "test_predictions": int(
            len(
                risk_test_df
            )
        ),
        "historical_error_rate_train": float(
            y_risk_train.mean()
        ),
        "historical_error_rate_test": float(
            y_risk_test.mean()
        ),
        "risk_features": "|".join(
            selected_risk_features
        ),
        **metrics,
    })

    # Logistic coefficients in standardized units.
    logistic_step = (
        error_risk_model
        .named_steps[
            "logistic"
        ]
    )

    coef_df = pd.DataFrame({
        "sport": sport,
        "feature": (
            selected_risk_features
        ),
        "standardized_coefficient": (
            logistic_step
            .coef_[0]
        ),
        "standardized_odds_ratio": np.exp(
            logistic_step
            .coef_[0]
        ),
    })

    coef_df[
        "increases_error_risk"
    ] = (
        coef_df[
            "standardized_coefficient"
        ]
        > 0
    )

    coefficient_frames.append(
        coef_df
    )

    safe_sport = (
        sport.lower()
        .replace(" ", "_")
    )

    training_path = (
        RESULTS_DIR
        / (
            f"week12_{safe_sport}_"
            "error_risk_training_dataset.csv"
        )
    )

    test_path = (
        RESULTS_DIR
        / (
            f"week12_{safe_sport}_"
            "error_risk_test_predictions.csv"
        )
    )

    flip_path = (
        RESULTS_DIR
        / (
            f"week12_{safe_sport}_"
            "fragility_flip_details.csv"
        )
    )

    risk_train_df.to_csv(
        training_path,
        index=False,
    )

    risk_test_df.to_csv(
        test_path,
        index=False,
    )

    combined_flip_details = pd.concat(
        [
            validation_flip_details.assign(
                ERROR_RISK_DATA_SPLIT=(
                    "ERROR_RISK_TRAIN"
                )
            ),
            test_flip_details.assign(
                ERROR_RISK_DATA_SPLIT=(
                    "ERROR_RISK_TEST"
                )
            ),
        ],
        ignore_index=True,
    )

    combined_flip_details[
        "sport"
    ] = sport

    combined_flip_details.to_csv(
        flip_path,
        index=False,
    )

    all_training_frames.append(
        risk_train_df
    )

    all_test_frames.append(
        risk_test_df
    )

    all_fragility_feature_frames.append(
        combined_flip_details
    )

    model_path = (
        MODELS_DIR
        / (
            f"{safe_sport}_"
            "error_risk_model.pkl"
        )
    )

    with model_path.open(
        "wb"
    ) as file:
        pickle.dump(
            {
                "model": error_risk_model,
                "risk_features": (
                    selected_risk_features
                ),
                "primary_winner_model": (
                    primary_model_name
                ),
            },
            file,
        )

    model_paths.append(
        model_path
    )

    figure_paths.append(
        save_confusion_matrix_figure(
            y_true=y_risk_test,
            probability=(
                test_error_probability
            ),
            sport=sport,
        )
    )

    figure_paths.append(
        save_roc_figure(
            y_true=y_risk_test,
            probability=(
                test_error_probability
            ),
            sport=sport,
        )
    )

    sport_outputs[sport] = {
        "risk_train": risk_train_df,
        "risk_test": risk_test_df,
        "risk_model": error_risk_model,
        "risk_features": (
            selected_risk_features
        ),
        "winner_models": (
            fitted_winner_models
        ),
    }

    print(
        "Primary winner model:",
        primary_model_name,
    )

    print(
        "Error-risk features:",
        selected_risk_features,
    )

    print(
        "VALIDATION historical prediction errors:",
        f"{y_risk_train.mean():.1%}",
    )

    print(
        "TEST historical prediction errors:",
        f"{y_risk_test.mean():.1%}",
    )

    print(
        "TEST Error-Risk ROC-AUC:",
        round(
            metrics[
                "roc_auc"
            ],
            4,
        ),
    )

    print(
        "TEST Error Recall:",
        round(
            metrics[
                "recall_error"
            ],
            4,
        ),
    )


# ------------------------------------------------------------
# 11. Save combined Week 12 results
# ------------------------------------------------------------

error_risk_metrics = pd.DataFrame(
    metrics_rows
)

error_risk_coefficients = pd.concat(
    coefficient_frames,
    ignore_index=True,
)

all_error_risk_training = pd.concat(
    all_training_frames,
    ignore_index=True,
)

all_error_risk_test = pd.concat(
    all_test_frames,
    ignore_index=True,
)

all_fragility_flip_details = pd.concat(
    all_fragility_feature_frames,
    ignore_index=True,
)

METRICS_PATH = (
    RESULTS_DIR
    / "week12_error_risk_metrics.csv"
)

COEFFICIENTS_PATH = (
    RESULTS_DIR
    / "week12_error_risk_coefficients.csv"
)

ALL_TRAINING_PATH = (
    RESULTS_DIR
    / "week12_all_sports_error_risk_training_dataset.csv"
)

ALL_TEST_PATH = (
    RESULTS_DIR
    / "week12_all_sports_error_risk_test_predictions.csv"
)

ALL_FLIP_DETAILS_PATH = (
    RESULTS_DIR
    / "week12_all_sports_fragility_flip_details.csv"
)

error_risk_metrics.to_csv(
    METRICS_PATH,
    index=False,
)

error_risk_coefficients.to_csv(
    COEFFICIENTS_PATH,
    index=False,
)

all_error_risk_training.to_csv(
    ALL_TRAINING_PATH,
    index=False,
)

all_error_risk_test.to_csv(
    ALL_TEST_PATH,
    index=False,
)

all_fragility_flip_details.to_csv(
    ALL_FLIP_DETAILS_PATH,
    index=False,
)

print("\nWeek 12 error-risk metrics:")
display(error_risk_metrics)

print("\nStandardized error-risk coefficients:")
display(
    error_risk_coefficients
    .sort_values(
        [
            "sport",
            "standardized_coefficient",
        ],
        ascending=[
            True,
            False,
        ],
    )
)


# ------------------------------------------------------------
# 12. Error-risk probability distributions
# ------------------------------------------------------------

distribution_paths = []

for sport in [
    "NBA",
    "NFL",
    "Premier League",
]:
    sport_df = all_error_risk_test[
        all_error_risk_test[
            "sport"
        ].eq(
            sport
        )
    ].copy()

    correct = sport_df[
        sport_df[
            "PRIMARY_MODEL_ERROR"
        ].eq(0)
    ][
        "ERROR_RISK_PROBABILITY"
    ]

    wrong = sport_df[
        sport_df[
            "PRIMARY_MODEL_ERROR"
        ].eq(1)
    ][
        "ERROR_RISK_PROBABILITY"
    ]

    fig, ax = plt.subplots(
        figsize=(8, 5)
    )

    ax.hist(
        correct,
        bins=np.linspace(
            0,
            1,
            16,
        ),
        alpha=0.6,
        label="Primary prediction correct",
    )

    ax.hist(
        wrong,
        bins=np.linspace(
            0,
            1,
            16,
        ),
        alpha=0.6,
        label="Primary prediction wrong",
    )

    ax.set_xlim(
        0,
        1,
    )

    ax.set_xlabel(
        "Predicted Error-Risk Probability"
    )

    ax.set_ylabel(
        "Number of Games"
    )

    ax.set_title(
        f"{sport}: Error-Risk Probability Distribution"
    )

    ax.legend()
    fig.tight_layout()

    safe_sport = (
        sport.lower()
        .replace(" ", "_")
    )

    output_path = (
        FIGURES_DIR
        / (
            f"week12_{safe_sport}_"
            "error_risk_distribution.png"
        )
    )

    fig.savefig(
        output_path,
        dpi=160,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

    distribution_paths.append(
        output_path
    )


# ------------------------------------------------------------
# 13. Error risk by Fragility category
# ------------------------------------------------------------

fragility_risk_summary = (
    all_error_risk_test
    .groupby(
        [
            "sport",
            "FRAGILITY_CATEGORY",
        ],
        observed=False,
    )
    .agg(
        games=(
            "game_id",
            "size",
        ),
        actual_error_rate=(
            "PRIMARY_MODEL_ERROR",
            "mean",
        ),
        mean_predicted_error_risk=(
            "ERROR_RISK_PROBABILITY",
            "mean",
        ),
        mean_fragility_score=(
            "FRAGILITY_SCORE",
            "mean",
        ),
    )
    .reset_index()
)

FRAGILITY_RISK_SUMMARY_PATH = (
    RESULTS_DIR
    / "week12_error_risk_by_fragility_category.csv"
)

fragility_risk_summary.to_csv(
    FRAGILITY_RISK_SUMMARY_PATH,
    index=False,
)

print(
    "\nError risk by Fragility category:"
)

display(
    fragility_risk_summary
)


# ------------------------------------------------------------
# 14. Mean TEST error-risk probability by sport
# ------------------------------------------------------------

sport_risk_summary = (
    all_error_risk_test
    .groupby(
        "sport"
    )
    .agg(
        games=(
            "game_id",
            "size",
        ),
        actual_primary_error_rate=(
            "PRIMARY_MODEL_ERROR",
            "mean",
        ),
        mean_predicted_error_risk=(
            "ERROR_RISK_PROBABILITY",
            "mean",
        ),
        mean_fragility_score=(
            "FRAGILITY_SCORE",
            "mean",
        ),
    )
    .reset_index()
)

SPORT_RISK_SUMMARY_PATH = (
    RESULTS_DIR
    / "week12_error_risk_summary_by_sport.csv"
)

sport_risk_summary.to_csv(
    SPORT_RISK_SUMMARY_PATH,
    index=False,
)

fig, ax = plt.subplots(
    figsize=(8, 5)
)

x = np.arange(
    len(
        sport_risk_summary
    )
)

width = 0.35

ax.bar(
    x - width / 2,
    sport_risk_summary[
        "actual_primary_error_rate"
    ],
    width=width,
    label="Actual error rate",
)

ax.bar(
    x + width / 2,
    sport_risk_summary[
        "mean_predicted_error_risk"
    ],
    width=width,
    label="Mean predicted error risk",
)

ax.set_xticks(
    x
)

ax.set_xticklabels(
    sport_risk_summary[
        "sport"
    ]
)

ax.set_ylim(
    0,
    1,
)

ax.set_xlabel(
    "Sport"
)

ax.set_ylabel(
    "Rate / Probability"
)

ax.set_title(
    "Week 12 Actual vs Predicted Error Risk"
)

ax.legend()
fig.tight_layout()

SPORT_RISK_FIG_PATH = (
    FIGURES_DIR
    / "week12_actual_vs_predicted_error_risk_by_sport.png"
)

fig.savefig(
    SPORT_RISK_FIG_PATH,
    dpi=160,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


# ------------------------------------------------------------
# 15. Save Week 12 methodology
# ------------------------------------------------------------

methodology = {
    "project": PROJECT_TITLE,
    "week": WEEK_NUMBER,
    "target_definition": {
        "PRIMARY_MODEL_CORRECT": (
            "1 if original predicted winner was correct, otherwise 0"
        ),
        "PRIMARY_MODEL_ERROR": (
            "1 if original predicted winner was wrong, otherwise 0"
        ),
    },
    "error_risk_model_target": (
        "PRIMARY_MODEL_ERROR"
    ),
    "primary_model_fit_split": (
        "TRAIN"
    ),
    "error_risk_model_fit_split": (
        "VALIDATION"
    ),
    "error_risk_model_evaluation_split": (
        "TEST"
    ),
    "error_risk_model": (
        "Balanced Logistic Regression"
    ),
    "candidate_risk_features": (
        RISK_FEATURE_CANDIDATES
    ),
    "fragility_features": (
        FRAGILITY_FEATURES
    ),
    "fragility_step_sd": STEP_SD,
    "fragility_max_search_sd": (
        MAX_FLIP_SD
    ),
    "model_disagreement_definition": (
        "Number of Elo, Logistic Regression, and Gradient Boosting "
        "winner predictions that disagree with the selected primary model."
    ),
    "team_strength_definition": (
        "Absolute difference in pre-game win percentage."
    ),
    "probability_note": (
        "Week 12 uses original primary-model probability for leakage-safe "
        "VALIDATION risk-model training. Week 10 calibrated probability "
        "remains a separate input for Week 13."
    ),
}

METHODOLOGY_PATH = (
    RESULTS_DIR
    / "week12_error_risk_methodology.json"
)

METHODOLOGY_PATH.write_text(
    json.dumps(
        methodology,
        indent=2,
    ),
    encoding="utf-8",
)


# ------------------------------------------------------------
# 16. Write Week 12 documentation
# ------------------------------------------------------------

result_lines = []

for row in error_risk_metrics.itertuples(
    index=False
):
    result_lines.append(
        (
            f"- {row.sport}: ROC-AUC {row.roc_auc:.3f}, "
            f"error recall {row.recall_error:.3f}, "
            f"F1 {row.f1_error:.3f}, "
            f"Brier score {row.brier_score:.3f}."
        )
    )

result_text = "\n".join(
    result_lines
)

documentation = f"""# Against All Odds — Week 12 Error-Risk Model

Generated: {datetime.now(timezone.utc).isoformat(timespec="seconds")}

## Week 12 objective

Week 12 created a second model whose job is not to predict the winner of a
game. Instead, it estimates how likely the original winner prediction is to
be wrong.

## Target variables

For each historical prediction:

- PRIMARY_MODEL_CORRECT = 1 when the original winner prediction was correct.
- PRIMARY_MODEL_CORRECT = 0 when the original winner prediction was wrong.
- PRIMARY_MODEL_ERROR = 1 - PRIMARY_MODEL_CORRECT.

The Week 12 model predicts PRIMARY_MODEL_ERROR.

## Risk-model inputs

The project uses information motivated by the capstone draft:

- Original win probability
- Fragility Score
- Elo difference
- Team-strength difference
- Recent-form difference
- Model disagreement
- Rest difference
- Home advantage

A PRIMARY_PREDICTS_HOME indicator is also retained as a useful home-context
variable. Zero-variance variables are automatically removed before fitting.

## Chronological workflow

- Primary winner models are fit on TRAIN.
- Historical error-risk examples are created from VALIDATION.
- The error-risk model is trained on VALIDATION.
- Final error-risk performance is evaluated on TEST.

The held-out TEST outcomes are never used to train the winner model or the
error-risk model.

## Error-risk model

Separate balanced Logistic Regression error-risk models are trained for the
NBA, NFL, and Premier League. Balanced class weights are used because incorrect
winner predictions may be less common than correct predictions.

## Week 12 results

{result_text}

## Interpretation

A larger ERROR_RISK_PROBABILITY means the system believes the original winner
prediction has a greater chance of being wrong. This probability does not
replace the winner prediction. It is an additional reliability measure that
will be combined with probability calibration, the Fragility Score, and model
disagreement during Week 13.

## Next step

Week 13 combines winner prediction, probability calibration, Fragility Score,
error-risk probability, and model disagreement to create the Against All Odds
Unpredictability Index and Predict / Caution / Abstain system.
"""

DOC_PATH = (
    DOCS_DIR
    / "week12_error_risk_model.md"
)

DOC_PATH.write_text(
    documentation,
    encoding="utf-8",
)

readme_text = """## Week 12 — Error-Risk Model

Created a second model that estimates how likely the original winner prediction
is to be wrong. Historical predictions were labeled as correct or incorrect,
and the error-risk model used factors such as original win probability,
Fragility Score, Elo difference, team-strength difference, recent form, model
disagreement, rest, and home context. Separate models were trained for the NBA,
NFL, and Premier League using validation data and evaluated on the held-out test
data. The resulting error-risk probabilities will be used in the Week 13
Against All Odds Unpredictability Index.
"""

README_ADDITION_PATH = (
    DOCS_DIR
    / "week12_README_addition.md"
)

README_ADDITION_PATH.write_text(
    readme_text,
    encoding="utf-8",
)

print("✅ Saved:", DOC_PATH)
print("✅ Saved:", README_ADDITION_PATH)
print("✅ Saved:", METHODOLOGY_PATH)


# ------------------------------------------------------------
# 17. Week 12 quality checks
# ------------------------------------------------------------

quality_checks = {
    "three_sports_completed": (
        set(
            error_risk_metrics[
                "sport"
            ]
        )
        == {
            "NBA",
            "NFL",
            "Premier League",
        }
    ),
    "validation_used_for_risk_training": (
        all_error_risk_training[
            "data_split"
        ].eq(
            "VALIDATION"
        ).all()
    ),
    "test_used_for_final_evaluation": (
        all_error_risk_test[
            "data_split"
        ].eq(
            "TEST"
        ).all()
    ),
    "primary_model_correct_binary": (
        all_error_risk_training[
            "PRIMARY_MODEL_CORRECT"
        ].isin(
            [0, 1]
        ).all()
        and all_error_risk_test[
            "PRIMARY_MODEL_CORRECT"
        ].isin(
            [0, 1]
        ).all()
    ),
    "primary_model_error_binary": (
        all_error_risk_training[
            "PRIMARY_MODEL_ERROR"
        ].isin(
            [0, 1]
        ).all()
        and all_error_risk_test[
            "PRIMARY_MODEL_ERROR"
        ].isin(
            [0, 1]
        ).all()
    ),
    "error_risk_probabilities_valid": (
        all_error_risk_test[
            "ERROR_RISK_PROBABILITY"
        ].between(
            0,
            1,
        ).all()
    ),
    "fragility_scores_valid": (
        all_error_risk_training[
            "FRAGILITY_SCORE"
        ].between(
            0,
            100,
        ).all()
        and all_error_risk_test[
            "FRAGILITY_SCORE"
        ].between(
            0,
            100,
        ).all()
    ),
    "model_disagreement_valid": (
        all_error_risk_training[
            "MODEL_DISAGREEMENT_COUNT"
        ].between(
            0,
            2,
        ).all()
        and all_error_risk_test[
            "MODEL_DISAGREEMENT_COUNT"
        ].between(
            0,
            2,
        ).all()
    ),
    "three_error_risk_models_saved": (
        len(
            model_paths
        )
        == 3
        and all(
            path.exists()
            for path in model_paths
        )
    ),
    "required_results_exist": all([
        METRICS_PATH.exists(),
        COEFFICIENTS_PATH.exists(),
        ALL_TRAINING_PATH.exists(),
        ALL_TEST_PATH.exists(),
        ALL_FLIP_DETAILS_PATH.exists(),
        FRAGILITY_RISK_SUMMARY_PATH.exists(),
        SPORT_RISK_SUMMARY_PATH.exists(),
        METHODOLOGY_PATH.exists(),
    ]),
    "documentation_exists": (
        DOC_PATH.exists()
        and README_ADDITION_PATH.exists()
    ),
}

week12_quality = pd.DataFrame([
    {
        "check": name,
        "passed": bool(value),
    }
    for name, value in quality_checks.items()
])

QUALITY_PATH = (
    RESULTS_DIR
    / "week12_quality_report.csv"
)

week12_quality.to_csv(
    QUALITY_PATH,
    index=False,
)

print("\nWeek 12 quality checks:")
display(week12_quality)

if not week12_quality[
    "passed"
].all():
    raise RuntimeError(
        "Week 12 quality checks failed."
    )

print("✅ All Week 12 quality checks passed")


# ------------------------------------------------------------
# 18. Package Week 12 GitHub outputs
# ------------------------------------------------------------

# Re-create path variables so the export section remains
# robust if it is rerun after a Colab runtime reset.

PROJECT_ROOT = choose_project_root()
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = PROJECT_ROOT / "figures" / "week12"
DOCS_DIR = PROJECT_ROOT / "docs"
MODELS_DIR = PROJECT_ROOT / "models" / "week12"

for folder in [
    RESULTS_DIR,
    FIGURES_DIR,
    DOCS_DIR,
    MODELS_DIR,
]:
    folder.mkdir(
        parents=True,
        exist_ok=True,
    )

runtime_files = []

runtime_files.extend(
    RESULTS_DIR.glob(
        "week12_*"
    )
)

runtime_files.extend(
    FIGURES_DIR.glob(
        "week12_*.png"
    )
)

runtime_files.extend(
    MODELS_DIR.glob(
        "*.pkl"
    )
)

for path in [
    DOC_PATH,
    README_ADDITION_PATH,
]:
    if path.exists():
        runtime_files.append(path)

runtime_files = list(
    dict.fromkeys(
        path
        for path in runtime_files
        if path.exists()
    )
)

if len(runtime_files) == 0:
    raise FileNotFoundError(
        "No Week 12 output files were found. "
        "Run the full Week 12 code first."
    )

push_list = """AGAINST ALL ODDS — WEEK 12 GITHUB PUSH LIST

notebooks/
  14_error_risk_model.ipynb

docs/
  week12_error_risk_model.md
  week12_README_addition.md

results/
  week12_error_risk_metrics.csv
  week12_error_risk_coefficients.csv
  week12_all_sports_error_risk_training_dataset.csv
  week12_all_sports_error_risk_test_predictions.csv
  week12_all_sports_fragility_flip_details.csv
  week12_nba_error_risk_training_dataset.csv
  week12_nfl_error_risk_training_dataset.csv
  week12_premier_league_error_risk_training_dataset.csv
  week12_nba_error_risk_test_predictions.csv
  week12_nfl_error_risk_test_predictions.csv
  week12_premier_league_error_risk_test_predictions.csv
  week12_nba_fragility_flip_details.csv
  week12_nfl_fragility_flip_details.csv
  week12_premier_league_fragility_flip_details.csv
  week12_error_risk_by_fragility_category.csv
  week12_error_risk_summary_by_sport.csv
  week12_error_risk_methodology.json
  week12_quality_report.csv

figures/week12/
  week12_nba_error_risk_confusion.png
  week12_nfl_error_risk_confusion.png
  week12_premier_league_error_risk_confusion.png
  week12_nba_error_risk_roc.png
  week12_nfl_error_risk_roc.png
  week12_premier_league_error_risk_roc.png
  week12_nba_error_risk_distribution.png
  week12_nfl_error_risk_distribution.png
  week12_premier_league_error_risk_distribution.png
  week12_actual_vs_predicted_error_risk_by_sport.png

models/week12/
  nba_error_risk_model.pkl
  nfl_error_risk_model.pkl
  premier_league_error_risk_model.pkl

Also update README.md using:
docs/week12_README_addition.md
"""

CHECKLIST_PATH = (
    PROJECT_ROOT
    / "WEEK12_GITHUB_PUSH_LIST.txt"
)

CHECKLIST_PATH.write_text(
    push_list,
    encoding="utf-8",
)

if running_in_colab():
    WEEK12_ZIP = Path(
        "/content/Against_All_Odds_Week12_GitHub_Upload.zip"
    )
else:
    WEEK12_ZIP = (
        PROJECT_ROOT
        / "Against_All_Odds_Week12_GitHub_Upload.zip"
    )

if WEEK12_ZIP.exists():
    WEEK12_ZIP.unlink()

with zipfile.ZipFile(
    WEEK12_ZIP,
    "w",
    compression=zipfile.ZIP_DEFLATED,
) as archive:
    for path in runtime_files:
        archive.write(
            path,
            arcname=str(
                path.relative_to(
                    PROJECT_ROOT
                )
            ),
        )

    archive.write(
        CHECKLIST_PATH,
        arcname="WEEK12_GITHUB_PUSH_LIST.txt",
    )

print("\n✅ Week 12 GitHub ZIP created:")
print(WEEK12_ZIP)

print(
    "ZIP size:",
    round(
        WEEK12_ZIP.stat().st_size
        / (1024 * 1024),
        2,
    ),
    "MB",
)


# ------------------------------------------------------------
# 19. Final status
# ------------------------------------------------------------

print("\n" + "=" * 78)
print("AGAINST ALL ODDS — WEEK 12 COMPLETE")
print("=" * 78)

print("✅ Historical primary predictions labeled correct / incorrect")
print("✅ PRIMARY_MODEL_CORRECT created")
print("✅ PRIMARY_MODEL_ERROR created")
print("✅ Original win probability included")
print("✅ Fragility Score included")
print("✅ Elo difference included")
print("✅ Team-strength difference included")
print("✅ Recent-form difference included")
print("✅ Model disagreement included")
print("✅ Rest difference included")
print("✅ Home context included")
print("✅ Separate NBA error-risk model trained")
print("✅ Separate NFL error-risk model trained")
print("✅ Separate Premier League error-risk model trained")
print("✅ Error-risk probabilities generated on TEST")
print("✅ Metrics, figures, coefficients, and datasets saved")
print("✅ Quality checks passed")
print("✅ GitHub upload ZIP created")

print("\nNext project stage:")
print(
    "Week 13 — Unpredictability Index and "
    "Predict / Caution / Abstain system"
)

if running_in_colab():
    print("\nStarting Week 12 ZIP download...")

    try:
        from google.colab import files

        files.download(
            str(WEEK12_ZIP)
        )

        print("✅ Download started.")

    except Exception as error:
        print(
            "Automatic download did not start."
        )

        print(
            "The ZIP is still saved here:"
        )

        print(WEEK12_ZIP)
        print("Error:", error)
```

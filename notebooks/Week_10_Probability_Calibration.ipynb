Week 10: Probability Calibration

```python

# ============================================================
# AGAINST ALL ODDS — WEEK 10: PROBABILITY CALIBRATION
# ============================================================
# Required by the capstone draft:
#   - Calibration curves
#   - Brier score
#   - Log loss
#   - Reliability tables
#   - Calibrated winner probabilities
#
# Method used:
#   - Recreate the Week 9 selected primary model for each sport
#   - Fit Platt scaling on VALIDATION
#   - Evaluate raw vs calibrated probabilities on TEST
#   - TEST is never used to fit the base model or calibrator
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import pickle
import sys
import zipfile
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.calibration import calibration_curve
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    roc_auc_score,
    brier_score_loss,
    log_loss,
)

try:
    from IPython.display import display
except ImportError:
    def display(obj):
        print(obj)

warnings.filterwarnings("ignore", category=FutureWarning)

PROJECT_TITLE = "Against All Odds"
WEEK_NUMBER = 10
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

GB_PARAMS = {
    "n_estimators": 150,
    "learning_rate": 0.05,
    "max_depth": 3,
    "min_samples_leaf": 10,
    "subsample": 0.90,
    "random_state": RANDOM_SEED,
}

CORE_FEATURES = [
    "ELO_DIFF",
    "WIN_PCT_DIFF",
    "FORM_DIFF",
    "POINT_GOAL_DIFF",
    "OFFENSE_DIFF",
    "DEFENSE_DIFF",
    "REST_DIFF",
    "SCHEDULE_STRENGTH_DIFF",
]

MODEL_READY_INPUTS = {
    "NBA": "nba_model_ready_2021_22_to_2025_26.csv",
    "NFL": "nfl_model_ready_2021_to_2025.csv",
    "Premier League": "premier_league_model_ready_2021_22_to_2025_26.csv",
}

WEEK9_SELECTION_FILE = "week9_best_primary_model_by_sport.csv"
N_RELIABILITY_BINS = 10

print("=" * 78)
print(f"{PROJECT_TITLE} — WEEK {WEEK_NUMBER}: PROBABILITY CALIBRATION")
print("=" * 78)
print("Calibration method: Platt scaling")
print("Calibration fit split: VALIDATION")
print("Evaluation split: TEST")


# ============================================================
# 1. PROJECT PATHS
# ============================================================

def running_in_colab():
    return "google.colab" in sys.modules


def choose_project_root():
    cwd = Path.cwd().resolve()
    candidates = [
        cwd if cwd.name.lower().replace("_", "-") in {"againstallodds", "against-all-odds"} else None,
        Path("/content/againstallodds"),
        Path("/content/Against-All-Odds"),
        Path("/content/drive/MyDrive/againstallodds"),
        Path("/content/drive/MyDrive/Against-All-Odds"),
    ]

    for candidate in candidates:
        if candidate is not None and candidate.exists():
            return candidate

    if running_in_colab():
        return Path("/content/againstallodds")

    return cwd / "againstallodds"


PROJECT_ROOT = choose_project_root()
MODEL_READY_DIR = PROJECT_ROOT / "data" / "model_ready"
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = PROJECT_ROOT / "figures" / "week10"
DOCS_DIR = PROJECT_ROOT / "docs"
MODELS_DIR = PROJECT_ROOT / "models" / "week10"

for folder in [MODEL_READY_DIR, RESULTS_DIR, FIGURES_DIR, DOCS_DIR, MODELS_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)


# ============================================================
# 2. FIND REQUIRED INPUT FILES
# ============================================================

def search_locations():
    locations = [
        PROJECT_ROOT,
        MODEL_READY_DIR,
        RESULTS_DIR,
        Path("/content"),
        Path("/content/week10_inputs"),
        Path.cwd().resolve(),
    ]

    for extra in [
        Path("/content/drive/MyDrive/againstallodds"),
        Path("/content/drive/MyDrive/Against-All-Odds"),
    ]:
        if extra.exists():
            locations.append(extra)

    return list(dict.fromkeys([p for p in locations if p.exists()]))


def find_exact_file(filename):
    for root in search_locations():
        direct = root / filename
        if direct.exists():
            return direct

    for root in search_locations():
        try:
            matches = list(root.rglob(filename))
        except (PermissionError, OSError):
            matches = []
        if matches:
            return matches[0]

    return None


def resolve_required_files():
    required = list(MODEL_READY_INPUTS.values()) + [WEEK9_SELECTION_FILE]
    found = {name: find_exact_file(name) for name in required}
    missing = [name for name, path in found.items() if path is None]

    if missing and running_in_colab():
        print("\nMissing Week 10 input files:")
        for name in missing:
            print(" -", name)

        print("\nUpload the missing CSVs or a ZIP containing them.")
        from google.colab import files

        uploaded = files.upload()
        upload_dir = Path("/content/week10_inputs")
        upload_dir.mkdir(parents=True, exist_ok=True)

        for name, blob in uploaded.items():
            destination = upload_dir / Path(name).name
            destination.write_bytes(blob)

            if zipfile.is_zipfile(destination):
                with zipfile.ZipFile(destination, "r") as archive:
                    archive.extractall(upload_dir)

        found = {name: find_exact_file(name) for name in required}

    missing = [name for name, path in found.items() if path is None]

    if missing:
        raise FileNotFoundError(
            "Week 10 cannot start because these files are missing:\n"
            + "\n".join(f" - {name}" for name in missing)
            + "\n\nRun Weeks 6 and 9 first or upload the missing files."
        )

    return found


FILES = resolve_required_files()

print("\n✅ Required Week 10 files found:")
for name, path in FILES.items():
    print(f" - {name}: {path}")


# ============================================================
# 3. LOAD MODEL-READY DATA
# ============================================================

datasets = {}

for sport, filename in MODEL_READY_INPUTS.items():
    df = pd.read_csv(
        FILES[filename],
        dtype={"game_id": "string"},
        parse_dates=["game_date"],
    )

    df["data_split"] = (
        df["data_split"]
        .astype("string")
        .str.upper()
        .str.strip()
    )

    df["HOME_WIN"] = pd.to_numeric(df["HOME_WIN"], errors="coerce")

    required = ["game_id", "game_date", "data_split", "HOME_WIN"] + CORE_FEATURES
    missing = [col for col in required if col not in df.columns]

    if missing:
        raise ValueError(f"{sport} is missing required columns: {missing}")

    for feature in CORE_FEATURES:
        df[feature] = pd.to_numeric(df[feature], errors="coerce")

    df = (
        df.dropna(subset=CORE_FEATURES + ["HOME_WIN"])
        .sort_values(["game_date", "game_id"], kind="stable")
        .reset_index(drop=True)
    )

    if not df["HOME_WIN"].isin([0, 1]).all():
        raise ValueError(f"{sport}: HOME_WIN must contain only 0 or 1.")

    datasets[sport] = df

    print(
        f"{sport}: {len(df):,} rows | "
        f"TRAIN={df['data_split'].eq('TRAIN').sum():,} | "
        f"VALIDATION={df['data_split'].eq('VALIDATION').sum():,} | "
        f"TEST={df['data_split'].eq('TEST').sum():,}"
    )


# ============================================================
# 4. LOAD WEEK 9 PRIMARY MODEL CHOICES
# ============================================================

week9_selection = pd.read_csv(FILES[WEEK9_SELECTION_FILE])

if not {"sport", "model"}.issubset(week9_selection.columns):
    raise ValueError(
        "week9_best_primary_model_by_sport.csv must contain 'sport' and 'model'."
    )

week9_selection["sport"] = week9_selection["sport"].astype("string").str.strip()
week9_selection["model"] = week9_selection["model"].astype("string").str.strip()

PRIMARY_MODEL_BY_SPORT = dict(
    zip(
        week9_selection["sport"],
        week9_selection["model"],
    )
)

expected_sports = {"NBA", "NFL", "Premier League"}

if set(PRIMARY_MODEL_BY_SPORT) != expected_sports:
    raise ValueError(
        "Week 9 model-selection file must contain NBA, NFL, and Premier League."
    )

print("\nWeek 9 primary models:")
for sport, model_name in PRIMARY_MODEL_BY_SPORT.items():
    print(f" - {sport}: {model_name}")


# ============================================================
# 5. RECREATE RAW PRIMARY-MODEL PROBABILITIES
# ============================================================

def elo_probability(elo_diff):
    elo_diff = np.asarray(elo_diff, dtype=float)
    return 1.0 / (1.0 + 10.0 ** (-elo_diff / 400.0))


def build_logistic_model():
    return Pipeline([
        ("scaler", StandardScaler()),
        (
            "logistic",
            LogisticRegression(
                max_iter=5000,
                random_state=RANDOM_SEED,
            ),
        ),
    ])


def build_gradient_boosting_model():
    return GradientBoostingClassifier(**GB_PARAMS)


def generate_raw_probabilities(sport, df, model_name):
    train = df[df["data_split"].eq("TRAIN")].copy()
    validation = df[df["data_split"].eq("VALIDATION")].copy()
    test = df[df["data_split"].eq("TEST")].copy()

    if train.empty or validation.empty or test.empty:
        raise ValueError(
            f"{sport}: TRAIN, VALIDATION, and TEST must all contain rows."
        )

    model_key = str(model_name).strip().lower()
    fitted_model = None

    if model_key == "logistic regression":
        fitted_model = build_logistic_model()
        fitted_model.fit(
            train[CORE_FEATURES],
            train["HOME_WIN"].astype(int),
        )
        validation_prob = fitted_model.predict_proba(validation[CORE_FEATURES])[:, 1]
        test_prob = fitted_model.predict_proba(test[CORE_FEATURES])[:, 1]

    elif model_key == "gradient boosting":
        fitted_model = build_gradient_boosting_model()
        fitted_model.fit(
            train[CORE_FEATURES],
            train["HOME_WIN"].astype(int),
        )
        validation_prob = fitted_model.predict_proba(validation[CORE_FEATURES])[:, 1]
        test_prob = fitted_model.predict_proba(test[CORE_FEATURES])[:, 1]

    elif model_key == "elo":
        validation_prob = elo_probability(validation["ELO_DIFF"])
        test_prob = elo_probability(test["ELO_DIFF"])

    else:
        raise ValueError(
            f"{sport}: unsupported primary model '{model_name}'. "
            "Expected Elo, Logistic Regression, or Gradient Boosting."
        )

    validation_out = validation[
        ["game_id", "game_date", "data_split", "HOME_WIN"]
    ].copy()

    test_out = test[
        ["game_id", "game_date", "data_split", "HOME_WIN"]
    ].copy()

    validation_out["RAW_HOME_WIN_PROB"] = np.clip(validation_prob, 1e-6, 1 - 1e-6)
    test_out["RAW_HOME_WIN_PROB"] = np.clip(test_prob, 1e-6, 1 - 1e-6)

    return validation_out, test_out, fitted_model


# ============================================================
# 6. PLATT-SCALING CALIBRATOR
# ============================================================

def probability_to_logit(probability):
    p = np.clip(np.asarray(probability, dtype=float), 1e-6, 1 - 1e-6)
    return np.log(p / (1.0 - p))


def fit_platt_calibrator(raw_probability, y_true):
    x = probability_to_logit(raw_probability).reshape(-1, 1)
    y = np.asarray(y_true, dtype=int)

    calibrator = LogisticRegression(
        C=1e6,
        solver="lbfgs",
        max_iter=5000,
        random_state=RANDOM_SEED,
    )

    calibrator.fit(x, y)
    return calibrator


def apply_platt_calibrator(calibrator, raw_probability):
    x = probability_to_logit(raw_probability).reshape(-1, 1)
    probability = calibrator.predict_proba(x)[:, 1]
    return np.clip(probability, 1e-6, 1 - 1e-6)


# ============================================================
# 7. CALIBRATION METRICS AND RELIABILITY TABLES
# ============================================================

def evaluate_probability_quality(y_true, probability):
    y = np.asarray(y_true, dtype=int)
    p = np.clip(np.asarray(probability, dtype=float), 1e-6, 1 - 1e-6)
    pred = (p >= 0.50).astype(int)

    return {
        "accuracy": accuracy_score(y, pred),
        "f1": f1_score(y, pred, zero_division=0),
        "roc_auc": roc_auc_score(y, p),
        "brier_score": brier_score_loss(y, p),
        "log_loss": log_loss(y, p, labels=[0, 1]),
    }


def make_reliability_table(y_true, probability, sport, probability_type):
    frame = pd.DataFrame({
        "actual": np.asarray(y_true, dtype=int),
        "probability": np.clip(np.asarray(probability, dtype=float), 0, 1),
    })

    edges = np.linspace(0, 1, N_RELIABILITY_BINS + 1)
    labels = [
        f"{int(edges[i] * 100):02d}-{int(edges[i + 1] * 100):02d}%"
        for i in range(N_RELIABILITY_BINS)
    ]

    frame["probability_bin"] = pd.cut(
        frame["probability"],
        bins=edges,
        labels=labels,
        include_lowest=True,
        right=True,
    )

    table = (
        frame.groupby("probability_bin", observed=False)
        .agg(
            games=("actual", "size"),
            mean_predicted_probability=("probability", "mean"),
            actual_home_win_rate=("actual", "mean"),
        )
        .reset_index()
    )

    table.insert(0, "probability_type", probability_type)
    table.insert(0, "sport", sport)

    table["absolute_calibration_gap"] = (
        table["mean_predicted_probability"]
        - table["actual_home_win_rate"]
    ).abs()

    total = table["games"].sum()

    if total > 0:
        ece = (
            table["absolute_calibration_gap"].fillna(0)
            * table["games"]
        ).sum() / total
    else:
        ece = np.nan

    table["weighted_expected_calibration_error"] = ece
    return table


def save_calibration_curve(y_true, raw_prob, calibrated_prob, sport):
    raw_obs, raw_pred = calibration_curve(
        y_true,
        raw_prob,
        n_bins=N_RELIABILITY_BINS,
        strategy="uniform",
    )

    cal_obs, cal_pred = calibration_curve(
        y_true,
        calibrated_prob,
        n_bins=N_RELIABILITY_BINS,
        strategy="uniform",
    )

    fig, ax = plt.subplots(figsize=(7, 6))

    ax.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Perfect calibration",
    )

    ax.plot(
        raw_pred,
        raw_obs,
        marker="o",
        label="Raw probability",
    )

    ax.plot(
        cal_pred,
        cal_obs,
        marker="o",
        label="Calibrated probability",
    )

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xlabel("Mean Predicted Home-Win Probability")
    ax.set_ylabel("Observed Home-Win Rate")
    ax.set_title(f"{sport}: Week 10 Calibration Curve")
    ax.legend()
    fig.tight_layout()

    safe_sport = sport.lower().replace(" ", "_")
    output_path = FIGURES_DIR / f"week10_{safe_sport}_calibration_curve.png"

    fig.savefig(
        output_path,
        dpi=160,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

    return output_path


# ============================================================
# 8. CALIBRATE EACH SPORT
# ============================================================

metric_rows = []
reliability_frames = []
parameter_rows = []
prediction_paths = []
calibration_figure_paths = []
calibrator_paths = []
base_model_paths = []
week10_predictions = {}

for sport, df in datasets.items():
    primary_model = PRIMARY_MODEL_BY_SPORT[sport]

    validation_out, test_out, fitted_base_model = generate_raw_probabilities(
        sport,
        df,
        primary_model,
    )

    calibrator = fit_platt_calibrator(
        validation_out["RAW_HOME_WIN_PROB"],
        validation_out["HOME_WIN"],
    )

    validation_out["CALIBRATED_HOME_WIN_PROB"] = apply_platt_calibrator(
        calibrator,
        validation_out["RAW_HOME_WIN_PROB"],
    )

    test_out["CALIBRATED_HOME_WIN_PROB"] = apply_platt_calibrator(
        calibrator,
        test_out["RAW_HOME_WIN_PROB"],
    )

    test_out["RAW_PREDICTION"] = (
        test_out["RAW_HOME_WIN_PROB"] >= 0.50
    ).astype("int8")

    test_out["CALIBRATED_PREDICTION"] = (
        test_out["CALIBRATED_HOME_WIN_PROB"] >= 0.50
    ).astype("int8")

    test_out["PRIMARY_MODEL"] = primary_model
    test_out["CALIBRATION_METHOD"] = "Platt scaling"
    test_out["CALIBRATION_FIT_SPLIT"] = "VALIDATION"
    test_out["CALIBRATION_EVALUATION_SPLIT"] = "TEST"

    week10_predictions[sport] = test_out.copy()

    raw_metrics = evaluate_probability_quality(
        test_out["HOME_WIN"],
        test_out["RAW_HOME_WIN_PROB"],
    )

    calibrated_metrics = evaluate_probability_quality(
        test_out["HOME_WIN"],
        test_out["CALIBRATED_HOME_WIN_PROB"],
    )

    metric_rows.append({
        "sport": sport,
        "primary_model": primary_model,
        "probability_type": "RAW",
        "calibration_method": "None",
        "base_model_fit_split": "TRAIN",
        "calibration_fit_split": "None",
        "evaluation_split": "TEST",
        "games": len(test_out),
        **raw_metrics,
    })

    metric_rows.append({
        "sport": sport,
        "primary_model": primary_model,
        "probability_type": "CALIBRATED",
        "calibration_method": "Platt scaling",
        "base_model_fit_split": "TRAIN",
        "calibration_fit_split": "VALIDATION",
        "evaluation_split": "TEST",
        "games": len(test_out),
        **calibrated_metrics,
    })

    reliability_frames.append(
        make_reliability_table(
            test_out["HOME_WIN"],
            test_out["RAW_HOME_WIN_PROB"],
            sport,
            "RAW",
        )
    )

    reliability_frames.append(
        make_reliability_table(
            test_out["HOME_WIN"],
            test_out["CALIBRATED_HOME_WIN_PROB"],
            sport,
            "CALIBRATED",
        )
    )

    calibration_figure_paths.append(
        save_calibration_curve(
            test_out["HOME_WIN"].astype(int),
            test_out["RAW_HOME_WIN_PROB"],
            test_out["CALIBRATED_HOME_WIN_PROB"],
            sport,
        )
    )

    safe_sport = sport.lower().replace(" ", "_")

    prediction_path = (
        RESULTS_DIR
        / f"week10_{safe_sport}_calibrated_test_predictions.csv"
    )

    test_out.to_csv(prediction_path, index=False)
    prediction_paths.append(prediction_path)

    calibrator_path = (
        MODELS_DIR
        / f"{safe_sport}_platt_calibrator.pkl"
    )

    with calibrator_path.open("wb") as file:
        pickle.dump(calibrator, file)

    calibrator_paths.append(calibrator_path)

    if fitted_base_model is not None:
        model_path = (
            MODELS_DIR
            / f"{safe_sport}_week10_primary_model.pkl"
        )

        with model_path.open("wb") as file:
            pickle.dump(fitted_base_model, file)

        base_model_paths.append(model_path)

    parameter_rows.append({
        "sport": sport,
        "primary_model": primary_model,
        "calibration_method": "Platt scaling",
        "calibrator_intercept": float(calibrator.intercept_[0]),
        "calibrator_slope": float(calibrator.coef_[0, 0]),
        "validation_calibration_games": len(validation_out),
        "test_evaluation_games": len(test_out),
    })

    print(f"\n✅ {sport}: {primary_model} calibrated")
    print(
        "Brier:",
        f"{raw_metrics['brier_score']:.4f}",
        "->",
        f"{calibrated_metrics['brier_score']:.4f}",
    )
    print(
        "Log loss:",
        f"{raw_metrics['log_loss']:.4f}",
        "->",
        f"{calibrated_metrics['log_loss']:.4f}",
    )


# ============================================================
# 9. SAVE METRICS / TABLES
# ============================================================

calibration_metrics_df = pd.DataFrame(metric_rows)
reliability_table = pd.concat(reliability_frames, ignore_index=True)
calibration_parameters = pd.DataFrame(parameter_rows)

CALIBRATION_METRICS_PATH = RESULTS_DIR / "week10_calibration_metrics.csv"
RELIABILITY_TABLE_PATH = RESULTS_DIR / "week10_reliability_tables.csv"
CALIBRATION_PARAMETERS_PATH = RESULTS_DIR / "week10_calibration_parameters.csv"

calibration_metrics_df.to_csv(CALIBRATION_METRICS_PATH, index=False)
reliability_table.to_csv(RELIABILITY_TABLE_PATH, index=False)
calibration_parameters.to_csv(CALIBRATION_PARAMETERS_PATH, index=False)

display(calibration_metrics_df)
display(reliability_table)
display(calibration_parameters)


# ============================================================
# 10. RAW VS CALIBRATED COMPARISON
# ============================================================

comparison_rows = []

for sport in ["NBA", "NFL", "Premier League"]:
    raw = calibration_metrics_df[
        calibration_metrics_df["sport"].eq(sport)
        & calibration_metrics_df["probability_type"].eq("RAW")
    ].iloc[0]

    cal = calibration_metrics_df[
        calibration_metrics_df["sport"].eq(sport)
        & calibration_metrics_df["probability_type"].eq("CALIBRATED")
    ].iloc[0]

    comparison_rows.append({
        "sport": sport,
        "primary_model": cal["primary_model"],
        "raw_brier_score": raw["brier_score"],
        "calibrated_brier_score": cal["brier_score"],
        "brier_change_calibrated_minus_raw": (
            cal["brier_score"] - raw["brier_score"]
        ),
        "raw_log_loss": raw["log_loss"],
        "calibrated_log_loss": cal["log_loss"],
        "log_loss_change_calibrated_minus_raw": (
            cal["log_loss"] - raw["log_loss"]
        ),
        "raw_roc_auc": raw["roc_auc"],
        "calibrated_roc_auc": cal["roc_auc"],
        "raw_accuracy": raw["accuracy"],
        "calibrated_accuracy": cal["accuracy"],
    })

raw_vs_calibrated = pd.DataFrame(comparison_rows)

RAW_VS_CALIBRATED_PATH = (
    RESULTS_DIR
    / "week10_raw_vs_calibrated_comparison.csv"
)

raw_vs_calibrated.to_csv(
    RAW_VS_CALIBRATED_PATH,
    index=False,
)

display(raw_vs_calibrated)


# ============================================================
# 11. BRIER / LOG-LOSS COMPARISON FIGURES
# ============================================================

sports = raw_vs_calibrated["sport"].tolist()
x = np.arange(len(sports))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(
    x - width / 2,
    raw_vs_calibrated["raw_brier_score"],
    width=width,
    label="Raw",
)

ax.bar(
    x + width / 2,
    raw_vs_calibrated["calibrated_brier_score"],
    width=width,
    label="Calibrated",
)

ax.set_xticks(x)
ax.set_xticklabels(sports)
ax.set_xlabel("Sport")
ax.set_ylabel("Brier Score")
ax.set_title("Week 10 Raw vs Calibrated Brier Score")
ax.legend()
fig.tight_layout()

BRIER_FIG_PATH = (
    FIGURES_DIR
    / "week10_brier_score_comparison.png"
)

fig.savefig(
    BRIER_FIG_PATH,
    dpi=160,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(
    x - width / 2,
    raw_vs_calibrated["raw_log_loss"],
    width=width,
    label="Raw",
)

ax.bar(
    x + width / 2,
    raw_vs_calibrated["calibrated_log_loss"],
    width=width,
    label="Calibrated",
)

ax.set_xticks(x)
ax.set_xticklabels(sports)
ax.set_xlabel("Sport")
ax.set_ylabel("Log Loss")
ax.set_title("Week 10 Raw vs Calibrated Log Loss")
ax.legend()
fig.tight_layout()

LOG_LOSS_FIG_PATH = (
    FIGURES_DIR
    / "week10_log_loss_comparison.png"
)

fig.savefig(
    LOG_LOSS_FIG_PATH,
    dpi=160,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


# ============================================================
# 12. DOCUMENTATION
# ============================================================

summary_lines = []

for row in raw_vs_calibrated.itertuples(index=False):
    summary_lines.append(
        f"- {row.sport} ({row.primary_model}): "
        f"Brier {row.raw_brier_score:.4f} -> {row.calibrated_brier_score:.4f}; "
        f"log loss {row.raw_log_loss:.4f} -> {row.calibrated_log_loss:.4f}."
    )

summary_text = "\n".join(summary_lines)

report_text = f"""# Against All Odds — Week 10 Probability Calibration

Generated: {datetime.now(timezone.utc).isoformat(timespec="seconds")}

## Objective

Week 10 determines whether the selected primary winner-model probabilities can
actually be trusted. A well-calibrated model should have observed outcomes that
approximately match predicted probabilities.

## Calibration design

- Primary model selection: Week 9
- Base model fit: TRAIN
- Calibration method: Platt scaling
- Calibration fit: VALIDATION
- Calibration evaluation: TEST

TEST was not used to fit the base model or the calibrator.

## Required Week 10 outputs

- Calibration curves
- Brier score
- Log loss
- Reliability tables
- Calibrated winner probabilities

## Raw versus calibrated TEST results

{summary_text}

Lower Brier score and lower log loss indicate better probability quality.

## Next step

Week 11 will use the calibrated winner probabilities in the Against All Odds
Fragility Score analysis.
"""

WEEK10_REPORT_PATH = (
    DOCS_DIR
    / "week10_probability_calibration.md"
)

WEEK10_REPORT_PATH.write_text(
    report_text,
    encoding="utf-8",
)

readme_text = """## Week 10 — Probability Calibration

Evaluated whether the selected primary winner-model probabilities were reliable
for the NBA, NFL, and Premier League. Platt scaling was fit on validation data
and evaluated on the held-out test data. Calibration curves, Brier score, log
loss, and reliability tables were created for each sport. The resulting
calibrated winner probabilities will be used in later fragility and uncertainty
analysis.
"""

WEEK10_README_PATH = (
    DOCS_DIR
    / "week10_README_addition.md"
)

WEEK10_README_PATH.write_text(
    readme_text,
    encoding="utf-8",
)


# ============================================================
# 13. QUALITY CHECKS
# ============================================================

quality_checks = {
    "three_sports_loaded": (
        set(datasets.keys())
        == {"NBA", "NFL", "Premier League"}
    ),
    "three_primary_models_loaded": (
        set(PRIMARY_MODEL_BY_SPORT.keys())
        == {"NBA", "NFL", "Premier League"}
    ),
    "six_metric_rows_created": (
        len(calibration_metrics_df) == 6
    ),
    "all_calibrated_probabilities_valid": all(
        df["CALIBRATED_HOME_WIN_PROB"].between(0, 1).all()
        for df in week10_predictions.values()
    ),
    "validation_used_for_calibration": all(
        df["CALIBRATION_FIT_SPLIT"].eq("VALIDATION").all()
        for df in week10_predictions.values()
    ),
    "test_used_for_evaluation": all(
        df["CALIBRATION_EVALUATION_SPLIT"].eq("TEST").all()
        for df in week10_predictions.values()
    ),
    "three_calibration_curves_created": (
        len(calibration_figure_paths) == 3
        and all(path.exists() for path in calibration_figure_paths)
    ),
    "required_outputs_exist": all([
        CALIBRATION_METRICS_PATH.exists(),
        RELIABILITY_TABLE_PATH.exists(),
        CALIBRATION_PARAMETERS_PATH.exists(),
        RAW_VS_CALIBRATED_PATH.exists(),
        WEEK10_REPORT_PATH.exists(),
        WEEK10_README_PATH.exists(),
    ]),
}

week10_quality = pd.DataFrame([
    {
        "check": name,
        "passed": bool(value),
    }
    for name, value in quality_checks.items()
])

WEEK10_QUALITY_PATH = (
    RESULTS_DIR
    / "week10_quality_report.csv"
)

week10_quality.to_csv(
    WEEK10_QUALITY_PATH,
    index=False,
)

display(week10_quality)

if not week10_quality["passed"].all():
    raise RuntimeError(
        "Week 10 quality checks failed."
    )

print("✅ All Week 10 quality checks passed")


# ============================================================
# 14. CREATE GITHUB UPLOAD ZIP
# ============================================================

# Recreate folder variables here so export stays robust
# if this block is rerun later.
PROJECT_ROOT = choose_project_root()
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = PROJECT_ROOT / "figures" / "week10"
DOCS_DIR = PROJECT_ROOT / "docs"
MODELS_DIR = PROJECT_ROOT / "models" / "week10"

runtime_files = []
runtime_files.extend(RESULTS_DIR.glob("week10_*"))
runtime_files.extend(FIGURES_DIR.glob("week10_*.png"))
runtime_files.extend(MODELS_DIR.glob("*.pkl"))

for path in [WEEK10_REPORT_PATH, WEEK10_README_PATH]:
    if path.exists():
        runtime_files.append(path)

runtime_files = list(
    dict.fromkeys(
        path
        for path in runtime_files
        if path.exists()
    )
)

if not runtime_files:
    raise FileNotFoundError(
        "No Week 10 outputs were found. Run the full Week 10 code first."
    )

push_list = """AGAINST ALL ODDS — WEEK 10 GITHUB PUSH LIST

notebooks/
  12_calibration.ipynb

docs/
  week10_probability_calibration.md
  week10_README_addition.md

results/
  week10_calibration_metrics.csv
  week10_reliability_tables.csv
  week10_calibration_parameters.csv
  week10_raw_vs_calibrated_comparison.csv
  week10_nba_calibrated_test_predictions.csv
  week10_nfl_calibrated_test_predictions.csv
  week10_premier_league_calibrated_test_predictions.csv
  week10_quality_report.csv

figures/week10/
  week10_nba_calibration_curve.png
  week10_nfl_calibration_curve.png
  week10_premier_league_calibration_curve.png
  week10_brier_score_comparison.png
  week10_log_loss_comparison.png

models/week10/
  nba_platt_calibrator.pkl
  nfl_platt_calibrator.pkl
  premier_league_platt_calibrator.pkl
  *_week10_primary_model.pkl
  Note: a primary-model pickle is not created when the selected model is Elo.

Also update README.md using:
docs/week10_README_addition.md
"""

CHECKLIST_PATH = (
    PROJECT_ROOT
    / "WEEK10_GITHUB_PUSH_LIST.txt"
)

CHECKLIST_PATH.write_text(
    push_list,
    encoding="utf-8",
)

WEEK10_ZIP = Path(
    "/content/Against_All_Odds_Week10_GitHub_Upload.zip"
)

if WEEK10_ZIP.exists():
    WEEK10_ZIP.unlink()

with zipfile.ZipFile(
    WEEK10_ZIP,
    "w",
    compression=zipfile.ZIP_DEFLATED,
) as archive:
    for path in runtime_files:
        archive.write(
            path,
            arcname=str(
                path.relative_to(PROJECT_ROOT)
            ),
        )

    archive.write(
        CHECKLIST_PATH,
        arcname="WEEK10_GITHUB_PUSH_LIST.txt",
    )

print("\n✅ Week 10 GitHub ZIP created:")
print(WEEK10_ZIP)
print(
    "ZIP size:",
    round(
        WEEK10_ZIP.stat().st_size
        / (1024 * 1024),
        2,
    ),
    "MB",
)


# ============================================================
# 15. FINAL STATUS / AUTO DOWNLOAD
# ============================================================

print("\n" + "=" * 78)
print("AGAINST ALL ODDS — WEEK 10 COMPLETE")
print("=" * 78)

print("✅ Week 9 primary models loaded")
print("✅ Raw probabilities generated")
print("✅ Platt calibration fit on VALIDATION")
print("✅ Calibration evaluated on TEST")
print("✅ Calibration curves created")
print("✅ Brier score calculated")
print("✅ Log loss calculated")
print("✅ Reliability tables created")
print("✅ Calibrated winner probabilities saved")
print("✅ GitHub upload ZIP created")

print("\nNext project stage:")
print("Week 11 — Against All Odds Fragility Score")

if running_in_colab():
    try:
        from google.colab import files

        print("\nStarting Week 10 ZIP download...")
        files.download(str(WEEK10_ZIP))
        print("✅ Download started.")

    except Exception as error:
        print("\nAutomatic download did not start.")
        print("The ZIP is still saved here:")
        print(WEEK10_ZIP)
        print("Error:", error)
```

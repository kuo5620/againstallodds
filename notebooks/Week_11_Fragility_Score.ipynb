```python

from pathlib import Path
from datetime import datetime, timezone
import json
import pickle
import sys
import zipfile
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier

try:
    from IPython.display import display
except ImportError:
    def display(obj):
        print(obj)

warnings.filterwarnings("ignore", category=FutureWarning)

PROJECT_TITLE = "Against All Odds"
WEEK_NUMBER = 11
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

CORE_FEATURES = [
    "ELO_DIFF",
    "WIN_PCT_DIFF",
    "FORM_DIFF",
    "POINT_GOAL_DIFF",
    "OFFENSE_DIFF",
    "DEFENSE_DIFF",
    "REST_DIFF",
    "SCHEDULE_STRENGTH_DIFF",
]

FRAGILITY_FEATURES = [
    "ELO_DIFF",
    "FORM_DIFF",
    "OFFENSE_DIFF",
    "DEFENSE_DIFF",
    "REST_DIFF",
]

MODEL_READY_INPUTS = {
    "NBA": "nba_model_ready_2021_22_to_2025_26.csv",
    "NFL": "nfl_model_ready_2021_to_2025.csv",
    "Premier League": "premier_league_model_ready_2021_22_to_2025_26.csv",
}

WEEK9_SELECTION_FILE = "week9_best_primary_model_by_sport.csv"

GB_PARAMS = {
    "n_estimators": 150,
    "learning_rate": 0.05,
    "max_depth": 3,
    "min_samples_leaf": 10,
    "subsample": 0.90,
    "random_state": RANDOM_SEED,
}

STEP_SD = 0.25
MAX_FLIP_SD = 4.0
PERTURBATION_STEPS = np.arange(
    STEP_SD,
    MAX_FLIP_SD + STEP_SD / 2,
    STEP_SD,
)

print("=" * 78)
print(f"{PROJECT_TITLE} — WEEK {WEEK_NUMBER}")
print("AGAINST ALL ODDS FRAGILITY SCORE")
print("=" * 78)
print("Fragility features:", FRAGILITY_FEATURES)
print("Step size:", STEP_SD, "standard deviations")
print("Maximum search:", MAX_FLIP_SD, "standard deviations")


def running_in_colab():
    return "google.colab" in sys.modules


def choose_project_root():
    cwd = Path.cwd().resolve()

    candidates = [
        cwd if cwd.name.lower().replace("_", "-") in {
            "againstallodds",
            "against-all-odds",
        } else None,
        Path("/content/againstallodds"),
        Path("/content/Against-All-Odds"),
        Path("/content/drive/MyDrive/againstallodds"),
        Path("/content/drive/MyDrive/Against-All-Odds"),
    ]

    for candidate in candidates:
        if candidate is not None and candidate.exists():
            return candidate

    if running_in_colab():
        return Path("/content/againstallodds")

    return cwd / "againstallodds"


PROJECT_ROOT = choose_project_root()
MODEL_READY_DIR = PROJECT_ROOT / "data" / "model_ready"
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = PROJECT_ROOT / "figures" / "week11"
DOCS_DIR = PROJECT_ROOT / "docs"
MODELS_DIR = PROJECT_ROOT / "models" / "week11"

for folder in [
    MODEL_READY_DIR,
    RESULTS_DIR,
    FIGURES_DIR,
    DOCS_DIR,
    MODELS_DIR,
]:
    folder.mkdir(parents=True, exist_ok=True)

print("\nProject root:", PROJECT_ROOT)


def search_locations():
    locations = [
        PROJECT_ROOT,
        MODEL_READY_DIR,
        RESULTS_DIR,
        Path("/content"),
        Path("/content/week11_inputs"),
        Path.cwd().resolve(),
    ]

    for extra in [
        Path("/content/drive/MyDrive/againstallodds"),
        Path("/content/drive/MyDrive/Against-All-Odds"),
    ]:
        if extra.exists():
            locations.append(extra)

    return list(
        dict.fromkeys(
            path
            for path in locations
            if path.exists()
        )
    )


def find_exact_file(filename):
    for root in search_locations():
        direct = root / filename

        if direct.exists():
            return direct

    for root in search_locations():
        try:
            matches = list(
                root.rglob(filename)
            )
        except (PermissionError, OSError):
            matches = []

        if matches:
            return matches[0]

    return None


def locate_required_files():
    required_names = (
        list(MODEL_READY_INPUTS.values())
        + [WEEK9_SELECTION_FILE]
    )

    found = {
        name: find_exact_file(name)
        for name in required_names
    }

    missing = [
        name
        for name, path in found.items()
        if path is None
    ]

    if missing and running_in_colab():
        print("\nMissing files needed for Week 11:")

        for name in missing:
            print(" -", name)

        print(
            "\nUpload the missing CSV files "
            "or a ZIP containing them."
        )

        from google.colab import files

        uploaded = files.upload()

        upload_dir = Path("/content/week11_inputs")
        upload_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        for name, blob in uploaded.items():
            destination = (
                upload_dir
                / Path(name).name
            )

            destination.write_bytes(blob)

            if zipfile.is_zipfile(destination):
                with zipfile.ZipFile(
                    destination,
                    "r",
                ) as archive:
                    archive.extractall(upload_dir)

        found = {
            name: find_exact_file(name)
            for name in required_names
        }

    still_missing = [
        name
        for name, path in found.items()
        if path is None
    ]

    if still_missing:
        raise FileNotFoundError(
            "Week 11 cannot start because these files are missing:\n"
            + "\n".join(
                f" - {name}"
                for name in still_missing
            )
            + "\n\nRun Week 6 and Week 9 first, "
              "or upload the missing files."
        )

    return found


LOCATED_FILES = locate_required_files()

print("\n✅ Required Week 11 files located:")

for name, path in LOCATED_FILES.items():
    print(" -", name, "->", path)


datasets = {}
required_columns = (
    [
        "game_id",
        "game_date",
        "data_split",
        "HOME_WIN",
    ]
    + CORE_FEATURES
)

for sport, filename in MODEL_READY_INPUTS.items():
    path = LOCATED_FILES[filename]

    df = pd.read_csv(
        path,
        dtype={"game_id": "string"},
        parse_dates=["game_date"],
    )

    df["data_split"] = (
        df["data_split"]
        .astype("string")
        .str.upper()
        .str.strip()
    )

    df["HOME_WIN"] = pd.to_numeric(
        df["HOME_WIN"],
        errors="coerce",
    )

    missing = [
        col
        for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"{sport} is missing required Week 11 columns: {missing}"
        )

    for feature in CORE_FEATURES:
        df[feature] = pd.to_numeric(
            df[feature],
            errors="coerce",
        )

    df = (
        df
        .dropna(
            subset=CORE_FEATURES + ["HOME_WIN"]
        )
        .sort_values(
            ["game_date", "game_id"],
            kind="stable",
        )
        .reset_index(drop=True)
    )

    if not df["HOME_WIN"].isin([0, 1]).all():
        raise ValueError(
            f"{sport}: HOME_WIN must contain only 0 or 1."
        )

    datasets[sport] = df

    print(
        f"{sport}: {len(df):,} rows | "
        f"TRAIN={int(df['data_split'].eq('TRAIN').sum()):,} | "
        f"VALIDATION={int(df['data_split'].eq('VALIDATION').sum()):,} | "
        f"TEST={int(df['data_split'].eq('TEST').sum()):,}"
    )


selection_path = LOCATED_FILES[
    WEEK9_SELECTION_FILE
]

week9_selection = pd.read_csv(
    selection_path
)

required_selection_columns = [
    "sport",
    "model",
]

missing_selection = [
    col
    for col in required_selection_columns
    if col not in week9_selection.columns
]

if missing_selection:
    raise ValueError(
        "week9_best_primary_model_by_sport.csv "
        f"is missing: {missing_selection}"
    )

week9_selection["sport"] = (
    week9_selection["sport"]
    .astype("string")
    .str.strip()
)

week9_selection["model"] = (
    week9_selection["model"]
    .astype("string")
    .str.strip()
)

PRIMARY_MODEL_BY_SPORT = {
    row.sport: row.model
    for row in week9_selection[
        ["sport", "model"]
    ].itertuples(index=False)
}

expected_sports = {
    "NBA",
    "NFL",
    "Premier League",
}

if set(PRIMARY_MODEL_BY_SPORT) != expected_sports:
    raise ValueError(
        "Week 9 model-selection file must contain "
        "NBA, NFL, and Premier League."
    )

print("\nWeek 9 selected primary models:")

for sport, model_name in PRIMARY_MODEL_BY_SPORT.items():
    print(f" - {sport}: {model_name}")


def elo_probability(elo_difference):
    elo_difference = np.asarray(
        elo_difference,
        dtype=float,
    )

    return (
        1.0
        / (
            1.0
            + 10.0 ** (
                -elo_difference
                / 400.0
            )
        )
    )


def build_logistic_model():
    return Pipeline([
        (
            "scaler",
            StandardScaler(),
        ),
        (
            "logistic",
            LogisticRegression(
                max_iter=5000,
                random_state=RANDOM_SEED,
            ),
        ),
    ])


def build_gradient_boosting_model():
    return GradientBoostingClassifier(
        **GB_PARAMS
    )


def probability_to_logit(probability):
    probability = np.clip(
        np.asarray(
            probability,
            dtype=float,
        ),
        1e-6,
        1 - 1e-6,
    )

    return np.log(
        probability
        / (
            1.0
            - probability
        )
    )


def fit_platt_calibrator(
    raw_probability,
    y_true,
):
    x = (
        probability_to_logit(
            raw_probability
        )
        .reshape(-1, 1)
    )

    calibrator = LogisticRegression(
        C=1e6,
        solver="lbfgs",
        max_iter=5000,
        random_state=RANDOM_SEED,
    )

    calibrator.fit(
        x,
        np.asarray(
            y_true,
            dtype=int,
        ),
    )

    return calibrator


def apply_platt_calibrator(
    calibrator,
    raw_probability,
):
    x = (
        probability_to_logit(
            raw_probability
        )
        .reshape(-1, 1)
    )

    probability = (
        calibrator
        .predict_proba(x)[:, 1]
    )

    return np.clip(
        probability,
        1e-6,
        1 - 1e-6,
    )


def fit_primary_model_and_calibrator(
    sport,
    df,
    model_name,
):
    train = df[
        df["data_split"].eq("TRAIN")
    ].copy()

    validation = df[
        df["data_split"].eq("VALIDATION")
    ].copy()

    test = df[
        df["data_split"].eq("TEST")
    ].copy()

    if train.empty:
        raise ValueError(
            f"{sport}: TRAIN split is empty."
        )

    if validation.empty:
        raise ValueError(
            f"{sport}: VALIDATION split is empty."
        )

    if test.empty:
        raise ValueError(
            f"{sport}: TEST split is empty."
        )

    model_key = (
        str(model_name)
        .strip()
        .lower()
    )

    fitted_model = None

    if model_key == "logistic regression":
        fitted_model = build_logistic_model()

        fitted_model.fit(
            train[CORE_FEATURES],
            train["HOME_WIN"].astype(int),
        )

        validation_raw = (
            fitted_model
            .predict_proba(
                validation[CORE_FEATURES]
            )[:, 1]
        )

    elif model_key == "gradient boosting":
        fitted_model = (
            build_gradient_boosting_model()
        )

        fitted_model.fit(
            train[CORE_FEATURES],
            train["HOME_WIN"].astype(int),
        )

        validation_raw = (
            fitted_model
            .predict_proba(
                validation[CORE_FEATURES]
            )[:, 1]
        )

    elif model_key == "elo":
        validation_raw = elo_probability(
            validation["ELO_DIFF"]
        )

    else:
        raise ValueError(
            f"{sport}: unsupported primary model '{model_name}'. "
            "Expected Elo, Logistic Regression, or Gradient Boosting."
        )

    calibrator = fit_platt_calibrator(
        validation_raw,
        validation["HOME_WIN"].astype(int),
    )

    return (
        train,
        validation,
        test,
        fitted_model,
        calibrator,
    )


def raw_probability_from_feature_frame(
    feature_frame,
    model_name,
    fitted_model,
):
    model_key = (
        str(model_name)
        .strip()
        .lower()
    )

    if model_key == "elo":
        return elo_probability(
            feature_frame["ELO_DIFF"]
        )

    if model_key in {
        "logistic regression",
        "gradient boosting",
    }:
        return (
            fitted_model
            .predict_proba(
                feature_frame[CORE_FEATURES]
            )[:, 1]
        )

    raise ValueError(
        f"Unsupported model: {model_name}"
    )


def calibrated_probability_from_feature_frame(
    feature_frame,
    model_name,
    fitted_model,
    calibrator,
):
    raw_probability = (
        raw_probability_from_feature_frame(
            feature_frame=feature_frame,
            model_name=model_name,
            fitted_model=fitted_model,
        )
    )

    calibrated_probability = (
        apply_platt_calibrator(
            calibrator,
            raw_probability,
        )
    )

    return (
        raw_probability,
        calibrated_probability,
    )


def fragility_category(score):
    if score >= 75:
        return "Extremely Fragile"

    if score >= 50:
        return "Fragile"

    if score >= 25:
        return "Moderate"

    return "Stable"


def analyze_fragility_for_sport(
    sport,
    df,
    primary_model_name,
):
    (
        train,
        validation,
        test,
        fitted_model,
        calibrator,
    ) = fit_primary_model_and_calibrator(
        sport=sport,
        df=df,
        model_name=primary_model_name,
    )

    (
        original_raw_probability,
        original_calibrated_probability,
    ) = (
        calibrated_probability_from_feature_frame(
            feature_frame=test,
            model_name=primary_model_name,
            fitted_model=fitted_model,
            calibrator=calibrator,
        )
    )

    original_prediction = (
        original_calibrated_probability
        >= 0.50
    ).astype(int)

    n_games = len(test)

    feature_scale = {}
    feature_min = {}
    feature_max = {}

    for feature in FRAGILITY_FEATURES:
        std_value = float(
            train[feature].std(
                ddof=0
            )
        )

        if (
            not np.isfinite(std_value)
            or std_value <= 0
        ):
            std_value = 1.0

        feature_scale[feature] = std_value
        feature_min[feature] = float(
            train[feature].min()
        )
        feature_max[feature] = float(
            train[feature].max()
        )

    feature_result_frames = []

    for feature in FRAGILITY_FEATURES:
        scale = feature_scale[
            feature
        ]

        best_distance = np.full(
            n_games,
            np.inf,
            dtype=float,
        )

        best_direction = np.full(
            n_games,
            "",
            dtype=object,
        )

        best_probability = np.full(
            n_games,
            np.nan,
            dtype=float,
        )

        best_raw_probability = np.full(
            n_games,
            np.nan,
            dtype=float,
        )

        best_perturbed_value = np.full(
            n_games,
            np.nan,
            dtype=float,
        )

        original_feature_value = (
            test[feature]
            .to_numpy(
                dtype=float
            )
        )

        for step_sd in PERTURBATION_STEPS:
            unresolved = (
                ~np.isfinite(
                    best_distance
                )
            )

            if not unresolved.any():
                break

            for direction_sign, direction_name in [
                (-1.0, "Decrease"),
                (1.0, "Increase"),
            ]:
                unresolved = (
                    ~np.isfinite(
                        best_distance
                    )
                )

                if not unresolved.any():
                    break

                perturbed = test.copy()

                proposed_value = (
                    original_feature_value
                    + (
                        direction_sign
                        * step_sd
                        * scale
                    )
                )

                proposed_value = np.clip(
                    proposed_value,
                    feature_min[feature],
                    feature_max[feature],
                )

                perturbed[
                    feature
                ] = proposed_value

                (
                    perturbed_raw,
                    perturbed_calibrated,
                ) = (
                    calibrated_probability_from_feature_frame(
                        feature_frame=perturbed,
                        model_name=primary_model_name,
                        fitted_model=fitted_model,
                        calibrator=calibrator,
                    )
                )

                perturbed_prediction = (
                    perturbed_calibrated
                    >= 0.50
                ).astype(int)

                flips = (
                    perturbed_prediction
                    != original_prediction
                )

                actual_standardized_distance = (
                    np.abs(
                        proposed_value
                        - original_feature_value
                    )
                    / scale
                )

                new_flip = (
                    flips
                    & unresolved
                    & (
                        actual_standardized_distance
                        > 0
                    )
                )

                if new_flip.any():
                    best_distance[
                        new_flip
                    ] = (
                        actual_standardized_distance[
                            new_flip
                        ]
                    )

                    best_direction[
                        new_flip
                    ] = direction_name

                    best_probability[
                        new_flip
                    ] = (
                        perturbed_calibrated[
                            new_flip
                        ]
                    )

                    best_raw_probability[
                        new_flip
                    ] = (
                        perturbed_raw[
                            new_flip
                        ]
                    )

                    best_perturbed_value[
                        new_flip
                    ] = (
                        proposed_value[
                            new_flip
                        ]
                    )

        feature_frame = pd.DataFrame({
            "sport": sport,
            "game_id": test[
                "game_id"
            ].astype("string").to_numpy(),
            "feature": feature,
            "original_feature_value": (
                original_feature_value
            ),
            "feature_train_std": scale,
            "flip_found": (
                np.isfinite(
                    best_distance
                )
            ),
            "flip_distance_sd": np.where(
                np.isfinite(
                    best_distance
                ),
                best_distance,
                np.nan,
            ),
            "flip_direction": best_direction,
            "perturbed_feature_value_at_flip": (
                best_perturbed_value
            ),
            "calibrated_probability_at_flip": (
                best_probability
            ),
            "raw_probability_at_flip": (
                best_raw_probability
            ),
        })

        feature_result_frames.append(
            feature_frame
        )

    feature_flip_results = pd.concat(
        feature_result_frames,
        ignore_index=True,
    )

    finite_only = (
        feature_flip_results[
            feature_flip_results[
                "flip_found"
            ]
        ]
        .sort_values(
            [
                "game_id",
                "flip_distance_sd",
                "feature",
            ]
        )
    )

    first_flip = (
        finite_only
        .groupby(
            "game_id",
            as_index=False,
        )
        .first()
        .rename(
            columns={
                "feature": "FRAGILITY_DRIVER",
                "flip_distance_sd": "MIN_FLIP_DISTANCE_SD",
                "flip_direction": "FLIP_DIRECTION",
                "calibrated_probability_at_flip": (
                    "CALIBRATED_PROBABILITY_AT_FLIP"
                ),
            }
        )
    )

    id_candidates = [
        "game_id",
        "game_date",
        "data_split",
        "HOME_WIN",
        "season",
        "home_team",
        "away_team",
        "home_team_abbr",
        "away_team_abbr",
    ]

    id_columns = [
        column
        for column in id_candidates
        if column in test.columns
    ]

    game_results = test[
        id_columns
    ].copy()

    game_results[
        "sport"
    ] = sport

    game_results[
        "PRIMARY_MODEL"
    ] = primary_model_name

    game_results[
        "ORIGINAL_RAW_HOME_WIN_PROB"
    ] = original_raw_probability

    game_results[
        "ORIGINAL_CALIBRATED_HOME_WIN_PROB"
    ] = original_calibrated_probability

    game_results[
        "ORIGINAL_PREDICTION"
    ] = original_prediction

    game_results[
        "ORIGINAL_CONFIDENCE"
    ] = np.maximum(
        original_calibrated_probability,
        1.0
        - original_calibrated_probability,
    )

    game_results[
        "DISTANCE_FROM_50_50"
    ] = np.abs(
        original_calibrated_probability
        - 0.50
    )

    if first_flip.empty:
        game_results[
            "FRAGILITY_DRIVER"
        ] = pd.NA

        game_results[
            "MIN_FLIP_DISTANCE_SD"
        ] = np.nan

        game_results[
            "FLIP_DIRECTION"
        ] = pd.NA

        game_results[
            "CALIBRATED_PROBABILITY_AT_FLIP"
        ] = np.nan

    else:
        game_results = (
            game_results
            .merge(
                first_flip[
                    [
                        "game_id",
                        "FRAGILITY_DRIVER",
                        "MIN_FLIP_DISTANCE_SD",
                        "FLIP_DIRECTION",
                        "CALIBRATED_PROBABILITY_AT_FLIP",
                    ]
                ],
                on="game_id",
                how="left",
            )
        )

    game_results[
        "FLIP_FOUND_WITHIN_RANGE"
    ] = (
        game_results[
            "MIN_FLIP_DISTANCE_SD"
        ].notna()
    )

    capped_distance = (
        game_results[
            "MIN_FLIP_DISTANCE_SD"
        ]
        .fillna(
            MAX_FLIP_SD
        )
        .clip(
            lower=0,
            upper=MAX_FLIP_SD,
        )
    )

    game_results[
        "FRAGILITY_SCORE"
    ] = (
        100.0
        * (
            1.0
            - (
                capped_distance
                / MAX_FLIP_SD
            )
        )
    ).clip(
        lower=0,
        upper=100,
    )

    game_results[
        "FRAGILITY_SCORE"
    ] = (
        game_results[
            "FRAGILITY_SCORE"
        ]
        .round(2)
    )

    game_results[
        "FRAGILITY_CATEGORY"
    ] = (
        game_results[
            "FRAGILITY_SCORE"
        ]
        .map(
            fragility_category
        )
    )

    game_results[
        "PREDICTION_CORRECT"
    ] = (
        game_results[
            "ORIGINAL_PREDICTION"
        ]
        == game_results[
            "HOME_WIN"
        ].astype(int)
    ).astype("int8")

    game_results[
        "PERTURBATION_STEP_SD"
    ] = STEP_SD

    game_results[
        "MAX_SEARCH_SD"
    ] = MAX_FLIP_SD

    game_results[
        "FRAGILITY_SCORE_METHOD"
    ] = (
        "100 * (1 - min(single-feature flip distance, 4 SD) / 4 SD)"
    )

    return {
        "game_results": game_results,
        "feature_flip_results": feature_flip_results,
        "train": train,
        "validation": validation,
        "test": test,
        "base_model": fitted_model,
        "calibrator": calibrator,
    }


sport_analysis = {}
all_game_results = []
all_feature_results = []

for sport, df in datasets.items():
    print("\n" + "=" * 70)
    print("Running Week 11 fragility analysis:", sport)
    print("=" * 70)

    result = analyze_fragility_for_sport(
        sport=sport,
        df=df,
        primary_model_name=(
            PRIMARY_MODEL_BY_SPORT[
                sport
            ]
        ),
    )

    sport_analysis[sport] = result

    game_results = result[
        "game_results"
    ]

    feature_results = result[
        "feature_flip_results"
    ]

    all_game_results.append(
        game_results
    )

    all_feature_results.append(
        feature_results
    )

    safe_sport = (
        sport.lower()
        .replace(" ", "_")
    )

    game_path = (
        RESULTS_DIR
        / (
            f"week11_{safe_sport}_"
            "fragility_scores.csv"
        )
    )

    feature_path = (
        RESULTS_DIR
        / (
            f"week11_{safe_sport}_"
            "feature_flip_analysis.csv"
        )
    )

    game_results.to_csv(
        game_path,
        index=False,
    )

    feature_results.to_csv(
        feature_path,
        index=False,
    )

    calibrator_path = (
        MODELS_DIR
        / (
            f"{safe_sport}_"
            "fragility_platt_calibrator.pkl"
        )
    )

    with calibrator_path.open(
        "wb"
    ) as file:
        pickle.dump(
            result["calibrator"],
            file,
        )

    if result[
        "base_model"
    ] is not None:
        model_path = (
            MODELS_DIR
            / (
                f"{safe_sport}_"
                "fragility_primary_model.pkl"
            )
        )

        with model_path.open(
            "wb"
        ) as file:
            pickle.dump(
                result["base_model"],
                file,
            )

    print(
        "Games analyzed:",
        len(game_results),
    )

    print(
        "Mean Fragility Score:",
        round(
            game_results[
                "FRAGILITY_SCORE"
            ].mean(),
            2,
        ),
    )

    print("Category counts:")

    display(
        game_results[
            "FRAGILITY_CATEGORY"
        ]
        .value_counts()
        .rename_axis("category")
        .reset_index(name="games")
    )


all_fragility_scores = pd.concat(
    all_game_results,
    ignore_index=True,
)

all_feature_flip_results = pd.concat(
    all_feature_results,
    ignore_index=True,
)


fragility_summary_rows = []

for sport in [
    "NBA",
    "NFL",
    "Premier League",
]:
    sport_df = all_fragility_scores[
        all_fragility_scores[
            "sport"
        ].eq(sport)
    ].copy()

    counts = (
        sport_df[
            "FRAGILITY_CATEGORY"
        ]
        .value_counts()
        .to_dict()
    )

    fragility_summary_rows.append({
        "sport": sport,
        "primary_model": (
            PRIMARY_MODEL_BY_SPORT[sport]
        ),
        "games": int(len(sport_df)),
        "mean_fragility_score": float(
            sport_df[
                "FRAGILITY_SCORE"
            ].mean()
        ),
        "median_fragility_score": float(
            sport_df[
                "FRAGILITY_SCORE"
            ].median()
        ),
        "mean_original_confidence": float(
            sport_df[
                "ORIGINAL_CONFIDENCE"
            ].mean()
        ),
        "percent_flip_found_within_4sd": float(
            100
            * sport_df[
                "FLIP_FOUND_WITHIN_RANGE"
            ].mean()
        ),
        "stable_games": int(
            counts.get("Stable", 0)
        ),
        "moderate_games": int(
            counts.get("Moderate", 0)
        ),
        "fragile_games": int(
            counts.get("Fragile", 0)
        ),
        "extremely_fragile_games": int(
            counts.get(
                "Extremely Fragile",
                0,
            )
        ),
        "prediction_accuracy": float(
            sport_df[
                "PREDICTION_CORRECT"
            ].mean()
        ),
    })


fragility_summary = pd.DataFrame(
    fragility_summary_rows
)

category_summary = (
    all_fragility_scores
    .groupby(
        [
            "sport",
            "FRAGILITY_CATEGORY",
        ],
        observed=False,
    )
    .agg(
        games=(
            "game_id",
            "size",
        ),
        mean_fragility_score=(
            "FRAGILITY_SCORE",
            "mean",
        ),
        mean_prediction_confidence=(
            "ORIGINAL_CONFIDENCE",
            "mean",
        ),
        prediction_accuracy=(
            "PREDICTION_CORRECT",
            "mean",
        ),
    )
    .reset_index()
)

feature_summary = (
    all_feature_flip_results[
        all_feature_flip_results[
            "flip_found"
        ]
    ]
    .groupby(
        [
            "sport",
            "feature",
        ]
    )
    .agg(
        games_flipped=(
            "game_id",
            "size",
        ),
        mean_flip_distance_sd=(
            "flip_distance_sd",
            "mean",
        ),
        median_flip_distance_sd=(
            "flip_distance_sd",
            "median",
        ),
    )
    .reset_index()
)

driver_summary = (
    all_fragility_scores[
        all_fragility_scores[
            "FRAGILITY_DRIVER"
        ].notna()
    ]
    .groupby(
        [
            "sport",
            "FRAGILITY_DRIVER",
        ]
    )
    .size()
    .reset_index(
        name="games_where_feature_was_first_to_flip"
    )
)

FRAGILITY_SUMMARY_PATH = (
    RESULTS_DIR
    / "week11_fragility_summary.csv"
)

CATEGORY_SUMMARY_PATH = (
    RESULTS_DIR
    / "week11_fragility_category_summary.csv"
)

FEATURE_SUMMARY_PATH = (
    RESULTS_DIR
    / "week11_fragility_feature_summary.csv"
)

DRIVER_SUMMARY_PATH = (
    RESULTS_DIR
    / "week11_fragility_driver_summary.csv"
)

ALL_SCORES_PATH = (
    RESULTS_DIR
    / "week11_all_sports_fragility_scores.csv"
)

ALL_FEATURES_PATH = (
    RESULTS_DIR
    / "week11_all_sports_feature_flip_analysis.csv"
)

fragility_summary.to_csv(
    FRAGILITY_SUMMARY_PATH,
    index=False,
)

category_summary.to_csv(
    CATEGORY_SUMMARY_PATH,
    index=False,
)

feature_summary.to_csv(
    FEATURE_SUMMARY_PATH,
    index=False,
)

driver_summary.to_csv(
    DRIVER_SUMMARY_PATH,
    index=False,
)

all_fragility_scores.to_csv(
    ALL_SCORES_PATH,
    index=False,
)

all_feature_flip_results.to_csv(
    ALL_FEATURES_PATH,
    index=False,
)

print("\nWeek 11 overall summary:")
display(fragility_summary)

print("\nFragility categories:")
display(category_summary)

print("\nFeature flip summary:")
display(feature_summary)


histogram_paths = []

for sport in [
    "NBA",
    "NFL",
    "Premier League",
]:
    sport_df = all_fragility_scores[
        all_fragility_scores[
            "sport"
        ].eq(sport)
    ]

    fig, ax = plt.subplots(
        figsize=(8, 5)
    )

    ax.hist(
        sport_df[
            "FRAGILITY_SCORE"
        ],
        bins=np.arange(
            0,
            105,
            5,
        ),
        edgecolor="black",
    )

    ax.set_xlim(0, 100)
    ax.set_xlabel("Fragility Score")
    ax.set_ylabel("Number of Games")

    ax.set_title(
        f"{sport}: Against All Odds Fragility Score"
    )

    fig.tight_layout()

    safe_sport = (
        sport.lower()
        .replace(" ", "_")
    )

    output_path = (
        FIGURES_DIR
        / (
            f"week11_{safe_sport}_"
            "fragility_distribution.png"
        )
    )

    fig.savefig(
        output_path,
        dpi=160,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

    histogram_paths.append(
        output_path
    )


category_order = [
    "Stable",
    "Moderate",
    "Fragile",
    "Extremely Fragile",
]

category_plot = (
    all_fragility_scores
    .groupby(
        [
            "sport",
            "FRAGILITY_CATEGORY",
        ],
        observed=False,
    )
    .size()
    .unstack(
        fill_value=0
    )
    .reindex(
        columns=category_order,
        fill_value=0,
    )
)

fig, ax = plt.subplots(
    figsize=(10, 6)
)

category_plot.plot(
    kind="bar",
    ax=ax,
)

ax.set_xlabel("Sport")
ax.set_ylabel("Number of Games")
ax.set_title(
    "Week 11 Fragility Categories by Sport"
)
ax.tick_params(
    axis="x",
    rotation=0,
)
ax.legend(
    title="Fragility Category"
)

fig.tight_layout()

CATEGORY_FIG_PATH = (
    FIGURES_DIR
    / "week11_fragility_categories_by_sport.png"
)

fig.savefig(
    CATEGORY_FIG_PATH,
    dpi=160,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


mean_plot = (
    fragility_summary[
        [
            "sport",
            "mean_fragility_score",
        ]
    ]
)

fig, ax = plt.subplots(
    figsize=(8, 5)
)

ax.bar(
    mean_plot["sport"],
    mean_plot[
        "mean_fragility_score"
    ],
)

ax.set_ylim(0, 100)
ax.set_xlabel("Sport")
ax.set_ylabel("Mean Fragility Score")
ax.set_title(
    "Average Fragility by Sport"
)

fig.tight_layout()

MEAN_FIG_PATH = (
    FIGURES_DIR
    / "week11_mean_fragility_by_sport.png"
)

fig.savefig(
    MEAN_FIG_PATH,
    dpi=160,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


if not driver_summary.empty:
    driver_pivot = (
        driver_summary
        .pivot(
            index="sport",
            columns="FRAGILITY_DRIVER",
            values=(
                "games_where_feature_was_first_to_flip"
            ),
        )
        .fillna(0)
    )

    fig, ax = plt.subplots(
        figsize=(11, 6)
    )

    driver_pivot.plot(
        kind="bar",
        ax=ax,
    )

    ax.set_xlabel("Sport")
    ax.set_ylabel("Games")
    ax.set_title(
        "Which Feature Most Easily Flipped the Prediction?"
    )
    ax.tick_params(
        axis="x",
        rotation=0,
    )
    ax.legend(
        title="Fragility Driver"
    )

    fig.tight_layout()

    DRIVER_FIG_PATH = (
        FIGURES_DIR
        / "week11_fragility_driver_by_sport.png"
    )

    fig.savefig(
        DRIVER_FIG_PATH,
        dpi=160,
        bbox_inches="tight",
    )

    plt.show()
    plt.close(fig)

else:
    DRIVER_FIG_PATH = None

    print(
        "No single-feature flips were found, "
        "so the fragility-driver chart was skipped."
    )


fig, ax = plt.subplots(
    figsize=(9, 6)
)

for sport in [
    "NBA",
    "NFL",
    "Premier League",
]:
    sport_df = all_fragility_scores[
        all_fragility_scores[
            "sport"
        ].eq(sport)
    ]

    ax.scatter(
        sport_df[
            "ORIGINAL_CONFIDENCE"
        ],
        sport_df[
            "FRAGILITY_SCORE"
        ],
        alpha=0.55,
        label=sport,
    )

ax.set_xlabel(
    "Original Model Confidence"
)
ax.set_ylabel(
    "Fragility Score"
)
ax.set_title(
    "Prediction Confidence vs Fragility"
)
ax.legend()

fig.tight_layout()

CONFIDENCE_FIG_PATH = (
    FIGURES_DIR
    / "week11_confidence_vs_fragility.png"
)

fig.savefig(
    CONFIDENCE_FIG_PATH,
    dpi=160,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)


methodology = {
    "project": PROJECT_TITLE,
    "week": WEEK_NUMBER,
    "fragility_features": FRAGILITY_FEATURES,
    "perturbation_step_standard_deviations": STEP_SD,
    "maximum_flip_search_standard_deviations": MAX_FLIP_SD,
    "score_formula": (
        "100 * (1 - min(single-feature flip distance, 4 SD) / 4 SD)"
    ),
    "category_thresholds": {
        "Stable": "0 <= score < 25",
        "Moderate": "25 <= score < 50",
        "Fragile": "50 <= score < 75",
        "Extremely Fragile": "75 <= score <= 100",
    },
    "base_model_fit_split": "TRAIN",
    "calibration_fit_split": "VALIDATION",
    "fragility_analysis_split": "TEST",
    "player_availability": (
        "Not included because the current Week 9 primary models "
        "do not contain a reliable player-availability feature."
    ),
}

METHODOLOGY_JSON_PATH = (
    RESULTS_DIR
    / "week11_fragility_methodology.json"
)

METHODOLOGY_JSON_PATH.write_text(
    json.dumps(
        methodology,
        indent=2,
    ),
    encoding="utf-8",
)


summary_lines = []

for row in fragility_summary.itertuples(
    index=False
):
    summary_lines.append(
        (
            f"- {row.sport}: mean Fragility Score "
            f"{row.mean_fragility_score:.2f}; "
            f"{row.extremely_fragile_games} extremely fragile, "
            f"{row.fragile_games} fragile, "
            f"{row.moderate_games} moderate, "
            f"{row.stable_games} stable."
        )
    )

summary_text = "\n".join(
    summary_lines
)

report_text = f"""# Against All Odds — Week 11 Fragility Score

Generated: {datetime.now(timezone.utc).isoformat(timespec="seconds")}

## Week 11 objective

Week 11 created the original Against All Odds Fragility Score.

The purpose of the score is to measure how easily a model's predicted winner
can change when important pre-game information is changed slightly.

## Variables tested

{chr(10).join(f"- {feature}" for feature in FRAGILITY_FEATURES)}

Player availability was not included because the current primary prediction
models do not contain a reliable player-availability feature.

## Method

For each TEST game:

1. Generate the original calibrated winner probability.
2. Record the original predicted winner.
3. Change one important pre-game variable at a time.
4. Move the variable upward and downward in 0.25-standard-deviation steps.
5. Recalculate the calibrated probability after every change.
6. Stop when the predicted winner changes.
7. Record the smallest standardized change that caused a winner flip.

The search stops at {MAX_FLIP_SD:.1f} standard deviations.

## Fragility Score

Fragility Score = 100 × (1 - capped minimum flip distance / {MAX_FLIP_SD:.1f})

A high score means that only a small change was needed to flip the prediction.
A low score means the prediction was harder to change.

## Categories

- Stable: 0–<25
- Moderate: 25–<50
- Fragile: 50–<75
- Extremely Fragile: 75–100

## Results

{summary_text}

## Data protection

The primary model is fit on TRAIN. Probability calibration is fit on
VALIDATION. Fragility is measured on TEST. Actual TEST outcomes are not used
to search for the counterfactual change that flips the predicted winner.

## Next step

Week 12 will use the winner prediction, calibrated probability, Fragility
Score, team-strength information, and other reliability information to build
the prediction Error-Risk Model.
"""

WEEK11_REPORT_PATH = (
    DOCS_DIR
    / "week11_fragility_score.md"
)

WEEK11_REPORT_PATH.write_text(
    report_text,
    encoding="utf-8",
)

readme_text = """## Week 11 — Against All Odds Fragility Score

Created the original Against All Odds Fragility Score for NBA, NFL, and
Premier League predictions. Important pre-game variables were changed in small,
standardized steps and each prediction was recalculated until the predicted
winner flipped. Games were classified as Stable, Moderate, Fragile, or
Extremely Fragile based on how little change was needed to reverse the
prediction. These results will be used as an input to the Week 12 Error-Risk
Model.
"""

WEEK11_README_PATH = (
    DOCS_DIR
    / "week11_README_addition.md"
)

WEEK11_README_PATH.write_text(
    readme_text,
    encoding="utf-8",
)

print("✅ Saved:", WEEK11_REPORT_PATH)
print("✅ Saved:", WEEK11_README_PATH)
print("✅ Saved:", METHODOLOGY_JSON_PATH)


valid_categories = {
    "Stable",
    "Moderate",
    "Fragile",
    "Extremely Fragile",
}

quality_checks = {
    "three_sports_analyzed": (
        set(
            all_fragility_scores[
                "sport"
            ].unique()
        )
        == {
            "NBA",
            "NFL",
            "Premier League",
        }
    ),
    "all_scores_between_0_and_100": (
        all_fragility_scores[
            "FRAGILITY_SCORE"
        ].between(
            0,
            100,
        ).all()
    ),
    "all_categories_valid": (
        set(
            all_fragility_scores[
                "FRAGILITY_CATEGORY"
            ].unique()
        )
        .issubset(
            valid_categories
        )
    ),
    "all_original_probabilities_valid": (
        all_fragility_scores[
            "ORIGINAL_CALIBRATED_HOME_WIN_PROB"
        ].between(
            0,
            1,
        ).all()
    ),
    "analysis_uses_test_split": (
        all_fragility_scores[
            "data_split"
        ].eq("TEST").all()
    ),
    "five_main_features_tested": (
        set(
            all_feature_flip_results[
                "feature"
            ].unique()
        )
        == set(
            FRAGILITY_FEATURES
        )
    ),
    "feature_rows_complete": (
        len(
            all_feature_flip_results
        )
        == (
            len(
                all_fragility_scores
            )
            * len(
                FRAGILITY_FEATURES
            )
        )
    ),
    "summary_files_exist": all([
        FRAGILITY_SUMMARY_PATH.exists(),
        CATEGORY_SUMMARY_PATH.exists(),
        FEATURE_SUMMARY_PATH.exists(),
        DRIVER_SUMMARY_PATH.exists(),
        ALL_SCORES_PATH.exists(),
        ALL_FEATURES_PATH.exists(),
        METHODOLOGY_JSON_PATH.exists(),
    ]),
    "documentation_exists": (
        WEEK11_REPORT_PATH.exists()
        and WEEK11_README_PATH.exists()
    ),
}

week11_quality = pd.DataFrame([
    {
        "check": name,
        "passed": bool(value),
    }
    for name, value
    in quality_checks.items()
])

WEEK11_QUALITY_PATH = (
    RESULTS_DIR
    / "week11_quality_report.csv"
)

week11_quality.to_csv(
    WEEK11_QUALITY_PATH,
    index=False,
)

print("\nWeek 11 quality checks:")
display(week11_quality)

if not week11_quality[
    "passed"
].all():
    raise RuntimeError(
        "Week 11 quality checks failed."
    )

print("✅ All Week 11 quality checks passed")


PROJECT_ROOT = choose_project_root()
RESULTS_DIR = PROJECT_ROOT / "results"
FIGURES_DIR = PROJECT_ROOT / "figures" / "week11"
DOCS_DIR = PROJECT_ROOT / "docs"
MODELS_DIR = PROJECT_ROOT / "models" / "week11"

for folder in [
    RESULTS_DIR,
    FIGURES_DIR,
    DOCS_DIR,
    MODELS_DIR,
]:
    folder.mkdir(
        parents=True,
        exist_ok=True,
    )

runtime_files = []

runtime_files.extend(
    RESULTS_DIR.glob(
        "week11_*"
    )
)

runtime_files.extend(
    FIGURES_DIR.glob(
        "week11_*.png"
    )
)

runtime_files.extend(
    MODELS_DIR.glob(
        "*.pkl"
    )
)

for path in [
    WEEK11_REPORT_PATH,
    WEEK11_README_PATH,
]:
    if path.exists():
        runtime_files.append(path)

runtime_files = list(
    dict.fromkeys(
        path
        for path in runtime_files
        if path.exists()
    )
)

if len(runtime_files) == 0:
    raise FileNotFoundError(
        "No Week 11 outputs were found. "
        "Run the Week 11 analysis first."
    )

push_list = """AGAINST ALL ODDS — WEEK 11 GITHUB PUSH LIST

notebooks/
  13_fragility_analysis.ipynb

docs/
  week11_fragility_score.md
  week11_README_addition.md

results/
  week11_nba_fragility_scores.csv
  week11_nfl_fragility_scores.csv
  week11_premier_league_fragility_scores.csv
  week11_nba_feature_flip_analysis.csv
  week11_nfl_feature_flip_analysis.csv
  week11_premier_league_feature_flip_analysis.csv
  week11_all_sports_fragility_scores.csv
  week11_all_sports_feature_flip_analysis.csv
  week11_fragility_summary.csv
  week11_fragility_category_summary.csv
  week11_fragility_feature_summary.csv
  week11_fragility_driver_summary.csv
  week11_fragility_methodology.json
  week11_quality_report.csv

figures/week11/
  week11_nba_fragility_distribution.png
  week11_nfl_fragility_distribution.png
  week11_premier_league_fragility_distribution.png
  week11_fragility_categories_by_sport.png
  week11_mean_fragility_by_sport.png
  week11_fragility_driver_by_sport.png
  week11_confidence_vs_fragility.png

models/week11/
  *_fragility_platt_calibrator.pkl
  *_fragility_primary_model.pkl
  (The primary-model pickle is absent when the selected model is Elo.)

Also update the main README.md using:
docs/week11_README_addition.md
"""

CHECKLIST_PATH = (
    PROJECT_ROOT
    / "WEEK11_GITHUB_PUSH_LIST.txt"
)

CHECKLIST_PATH.write_text(
    push_list,
    encoding="utf-8",
)

if running_in_colab():
    WEEK11_ZIP = Path(
        "/content/Against_All_Odds_Week11_GitHub_Upload.zip"
    )
else:
    WEEK11_ZIP = (
        PROJECT_ROOT
        / "Against_All_Odds_Week11_GitHub_Upload.zip"
    )

if WEEK11_ZIP.exists():
    WEEK11_ZIP.unlink()

with zipfile.ZipFile(
    WEEK11_ZIP,
    "w",
    compression=zipfile.ZIP_DEFLATED,
) as archive:
    for path in runtime_files:
        archive.write(
            path,
            arcname=str(
                path.relative_to(
                    PROJECT_ROOT
                )
            ),
        )

    archive.write(
        CHECKLIST_PATH,
        arcname="WEEK11_GITHUB_PUSH_LIST.txt",
    )

print("\n✅ Week 11 GitHub ZIP created:")
print(WEEK11_ZIP)

print(
    "ZIP size:",
    round(
        WEEK11_ZIP.stat().st_size
        / (1024 * 1024),
        2,
    ),
    "MB",
)

print("\n" + "=" * 78)
print("AGAINST ALL ODDS — WEEK 11 COMPLETE")
print("=" * 78)

print("✅ Original calibrated prediction generated for every TEST game")
print("✅ Elo difference tested")
print("✅ Recent-form difference tested")
print("✅ Offensive difference tested")
print("✅ Defensive difference tested")
print("✅ Rest difference tested")
print("✅ Prediction recalculated after each change")
print("✅ Smallest winner-flip distance calculated")
print("✅ Standardized Fragility Score created")
print("✅ Stable / Moderate / Fragile / Extremely Fragile categories created")
print("✅ NBA Fragility Scores saved")
print("✅ NFL Fragility Scores saved")
print("✅ Premier League Fragility Scores saved")
print("✅ Summary tables created")
print("✅ Figures created")
print("✅ Documentation created")
print("✅ Quality checks passed")
print("✅ GitHub upload ZIP created")

print("\nNext project stage:")
print("Week 12 — Prediction Error-Risk Model")

if running_in_colab():
    print("\nStarting Week 11 ZIP download...")

    try:
        from google.colab import files

        files.download(
            str(WEEK11_ZIP)
        )

        print("✅ Download started.")

    except Exception as error:
        print(
            "Automatic download did not start."
        )

        print(
            "The ZIP is still saved here:"
        )

        print(WEEK11_ZIP)
        print("Error:", error)
```
